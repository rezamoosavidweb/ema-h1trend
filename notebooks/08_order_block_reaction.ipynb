# Notebook #3 — Order Block Reaction Strategy
## استراتژی واکنش به Order Block — XAUUSD (M5 + H1)

---

### تعریف Order Block
**Order Block (OB)** = آخرین کندل مخالف قبل از یک حرکت قوی

- **Bullish OB**: آخرین کندل **نزولی** قبل از یک حرکت صعودی قوی
- **Bearish OB**: آخرین کندل **صعودی** قبل از یک حرکت نزولی قوی

### چرا Order Block اهمیت دارد؟
بانک‌ها و موسسات مالی بزرگ در این نقاط سفارشات کلان خود را قرار می‌دهند.
وقتی قیمت به OB برمی‌گردد (Retest)، این موسسات مجدداً دست به کار می‌شوند.

### منطق استراتژی
1. **Displacement**: شناسایی حرکت قوی (≥ N کندل یک‌طرفه)
2. **Order Block**: آخرین کندل مخالف قبل از Displacement
3. **BOS** (Break of Structure): تأیید شکست ساختار
4. **FVG** (Fair Value Gap): محاسبه Gap در داخل Displacement
5. **Retest**: قیمت به OB برمی‌گردد
6. **Rejection Candle**: کندل تأییدی در OB
7. **Entry**: ورود بعد از Rejection

---

| پارامتر | مقدار |
|---|---|
| نماد | XAUUSD |
| تایم‌فریم | M5 |
| Displacement Candles | ≥ 4 |
| Displacement Strength | body ≥ 1.5× ATR |
| Risk/Reward | 1:2 |

## Step 1 — Imports & Configuration

````python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
from pathlib import Path
from typing import Optional, List, Tuple
from dataclasses import dataclass, field

import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio

pio.renderers.default = 'notebook'
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.4f}'.format)

# ── Config ───────────────────────────────────────────────────────────────────
SYMBOL     = 'XAUUSD'
DATA_DIR   = Path('./data')
RESULT_DIR = Path('./results')

# ── Date Range ───────────────────────────────────────────────────────────────
# Option 1: explicit date range  →  set DATE_FROM and/or DATE_TO
# Option 2: lookback from today  →  set DATE_FROM = None, use LOOKBACK_DAYS
DATE_FROM     = '2023-04-18'          # e.g. '2023-01-01'  or  None
DATE_TO       = '2026-05-21'          # e.g. '2024-01-01'  or  None (= today)
LOOKBACK_DAYS = 30           # used only when DATE_FROM is None

# ── Order Block Detection ────────────────────────────────────────────────────
DISPLACEMENT_MIN_CANDLES = 4     # min consecutive candles in one direction
DISPLACEMENT_ATR_MULT    = 1.5   # displacement move >= N * ATR
ATR_PERIOD               = 14
OB_MAX_LOOKBACK_BARS     = 200   # max bars back to look for valid OB
OB_EXPIRY_BARS           = 100   # OB expires after N bars without retest

# ── Entry Conditions ─────────────────────────────────────────────────────────
REJECTION_WICK_RATIO = 0.3   # wick must be > 30% of candle range
CONFIRM_CANDLES      = 1     # candles confirming OB reaction

# ── Risk ─────────────────────────────────────────────────────────────────────
RISK_REWARD    = 2
SL_BUFFER      = 0.5
MAX_TRADE_BARS = 288   # 24h in M5

print('Order Block Strategy — Config loaded.')
if DATE_FROM or DATE_TO:
    print(f'  Date range : {DATE_FROM or "start"} → {DATE_TO or "today"}')
else:
    print(f'  Lookback   : {LOOKBACK_DAYS} days from today')
````

## Step 2 — Load Data

````python
def load_ohlcv(symbol: str, tf: str,
               date_from=DATE_FROM, date_to=DATE_TO,
               lookback_days: int = LOOKBACK_DAYS) -> pd.DataFrame:
    path = DATA_DIR / symbol / tf / 'ohlcv.csv'
    if not path.exists():
        raise FileNotFoundError(f'Missing: {path}')
    df = pd.read_csv(path)
    df['time'] = pd.to_datetime(df['time'], utc=True)
    df = df.sort_values('time').reset_index(drop=True)
    keep = ['time', 'open', 'high', 'low', 'close', 'tick_volume']
    df = df[[c for c in keep if c in df.columns]].copy()
    df.rename(columns={'tick_volume': 'volume'}, inplace=True)

    # ── Filter by date range or lookback ────────────────────────────────────
    if date_from is not None:
        start = pd.Timestamp(date_from, tz='UTC')
    else:
        start = pd.Timestamp.now(tz='UTC') - pd.Timedelta(days=lookback_days)
    df = df[df['time'] >= start]

    if date_to is not None:
        end = pd.Timestamp(date_to, tz='UTC')
        df = df[df['time'] <= end]

    return df.copy().reset_index(drop=True)


df_m5 = load_ohlcv(SYMBOL, 'M5')

# Compute actual backtest window for downstream use
BACKTEST_START = df_m5['time'].min()
BACKTEST_END   = df_m5['time'].max()
BACKTEST_DAYS  = (BACKTEST_END - BACKTEST_START).days

print(f'M5 bars : {len(df_m5):,}')
print(f'Range   : {BACKTEST_START.date()} → {BACKTEST_END.date()}  ({BACKTEST_DAYS} days)')
display(df_m5.tail(3))
````

## Step 3 — ATR & Candle Analysis

````python
def add_candle_features(df: pd.DataFrame, atr_period: int = ATR_PERIOD) -> pd.DataFrame:
    """Add ATR, body, wick metrics for displacement and OB detection."""
    df = df.copy()

    # True Range and ATR
    df['tr'] = np.maximum(
        df['high'] - df['low'],
        np.maximum(
            abs(df['high'] - df['close'].shift(1)),
            abs(df['low']  - df['close'].shift(1))
        )
    )
    df['atr'] = df['tr'].ewm(span=atr_period, adjust=False).mean()

    # Candle body metrics
    df['body']       = abs(df['close'] - df['open'])
    df['body_top']   = df[['open', 'close']].max(axis=1)
    df['body_bot']   = df[['open', 'close']].min(axis=1)
    df['upper_wick'] = df['high'] - df['body_top']
    df['lower_wick'] = df['body_bot'] - df['low']
    df['range']      = df['high'] - df['low']
    df['is_bull']    = df['close'] > df['open']
    df['is_bear']    = df['close'] < df['open']

    return df


df_m5 = add_candle_features(df_m5)
print('ATR and candle features added.')
print(f'  ATR (current)   : {df_m5["atr"].iloc[-1]:.2f}')
print(f'  Avg Body        : {df_m5["body"].mean():.2f}')
print(f'  Avg Range       : {df_m5["range"].mean():.2f}')
````

## Step 4 — Displacement Detection

**Displacement** = حرکت قوی یک‌طرفه که نشان‌دهنده حضور موسسات مالی است.

شرایط Displacement:
- حداقل N کندل متوالی در یک جهت
- مجموع حرکت ≥ ATR × multiplier
- کندل‌های کمی wick و بیشتر body (مومنتوم واقعی)

````python
@dataclass
class Displacement:
    start_idx: int
    end_idx: int
    direction: str
    total_move: float
    candle_count: int
    start_price: float
    end_price: float
    start_time: object
    end_time: object


def detect_displacements(df, min_candles=DISPLACEMENT_MIN_CANDLES, atr_mult=DISPLACEMENT_ATR_MULT):
    displacements = []
    n = len(df)
    i = 0
    while i < n:
        if df.iloc[i]['is_bull']:
            j = i + 1
            while j < n and df.iloc[j]['is_bull']:
                j += 1
            count = j - i
            if count >= min_candles:
                segment = df.iloc[i:j]
                total_move = segment['close'].iloc[-1] - segment['open'].iloc[0]
                avg_atr = segment['atr'].mean()
                if total_move >= atr_mult * avg_atr:
                    displacements.append(Displacement(
                        start_idx=i, end_idx=j - 1, direction='UP',
                        total_move=total_move, candle_count=count,
                        start_price=segment['open'].iloc[0],
                        end_price=segment['close'].iloc[-1],
                        start_time=segment['time'].iloc[0],
                        end_time=segment['time'].iloc[-1],
                    ))
            i = j
        elif df.iloc[i]['is_bear']:
            j = i + 1
            while j < n and df.iloc[j]['is_bear']:
                j += 1
            count = j - i
            if count >= min_candles:
                segment = df.iloc[i:j]
                total_move = segment['open'].iloc[0] - segment['close'].iloc[-1]
                avg_atr = segment['atr'].mean()
                if total_move >= atr_mult * avg_atr:
                    displacements.append(Displacement(
                        start_idx=i, end_idx=j - 1, direction='DOWN',
                        total_move=total_move, candle_count=count,
                        start_price=segment['open'].iloc[0],
                        end_price=segment['close'].iloc[-1],
                        start_time=segment['time'].iloc[0],
                        end_time=segment['time'].iloc[-1],
                    ))
            i = j
        else:
            i += 1
    return displacements


displacements = detect_displacements(df_m5)

up_d   = [d for d in displacements if d.direction == 'UP']
down_d = [d for d in displacements if d.direction == 'DOWN']

print(f'Total displacements: {len(displacements)}')
print(f'  UP   : {len(up_d)}')
print(f'  DOWN : {len(down_d)}')
if displacements:
    moves = [d.total_move for d in displacements]
    print(f'  Avg move  : {np.mean(moves):.2f}')
    print(f'  Max move  : {max(moves):.2f}')
````

## Step 5 — Order Block Identification

پس از شناسایی هر Displacement:
- **Bullish OB**: آخرین کندل **نزولی** قبل از Displacement UP
- **Bearish OB**: آخرین کندل **صعودی** قبل از Displacement DOWN

```
Bullish OB:
  [bear][bear][BULL OB ← last bear before UP]
  ↓ Displacement UP →
  [bull][bull][bull][bull]

قیمت به OB برمی‌گردد → BUY entry
```

````python
@dataclass
class OrderBlock:
    disp_idx:   int         # index of displacement this OB belongs to
    ob_bar_idx: int         # bar index of the OB candle
    ob_type:    str         # 'bullish' or 'bearish'
    ob_high:    float
    ob_low:     float
    ob_time:    object
    disp_dir:   str         # 'UP' or 'DOWN'
    displaced_by: float     # magnitude of displacement move
    is_mitigated: bool = False
    is_expired:   bool = False


def find_order_blocks(df: pd.DataFrame, displacements: List[Displacement]) -> List[OrderBlock]:
    """
    For each displacement, find the last opposite candle before it starts.
    That candle = Order Block.
    """
    order_blocks = []

    for disp_num, disp in enumerate(displacements):
        start_i = disp.start_idx

        if disp.direction == 'UP':
            # Bullish OB = last bearish candle before displacement start
            look_back = max(0, start_i - 20)
            for k in range(start_i - 1, look_back - 1, -1):
                if df.iloc[k]['is_bear']:
                    bar = df.iloc[k]
                    order_blocks.append(OrderBlock(
                        disp_idx   = disp_num,
                        ob_bar_idx = k,
                        ob_type    = 'bullish',
                        ob_high    = bar['high'],
                        ob_low     = bar['low'],
                        ob_time    = bar['time'],
                        disp_dir   = 'UP',
                        displaced_by = disp.total_move,
                    ))
                    break

        else:  # DOWN
            # Bearish OB = last bullish candle before displacement start
            look_back = max(0, start_i - 20)
            for k in range(start_i - 1, look_back - 1, -1):
                if df.iloc[k]['is_bull']:
                    bar = df.iloc[k]
                    order_blocks.append(OrderBlock(
                        disp_idx   = disp_num,
                        ob_bar_idx = k,
                        ob_type    = 'bearish',
                        ob_high    = bar['high'],
                        ob_low     = bar['low'],
                        ob_time    = bar['time'],
                        disp_dir   = 'DOWN',
                        displaced_by = disp.total_move,
                    ))
                    break

    return order_blocks


order_blocks = find_order_blocks(df_m5, displacements)

bull_obs = [ob for ob in order_blocks if ob.ob_type == 'bullish']
bear_obs = [ob for ob in order_blocks if ob.ob_type == 'bearish']

print(f'Order Blocks found: {len(order_blocks)}')
print(f'  Bullish OB (→ BUY): {len(bull_obs)}')
print(f'  Bearish OB (→ SELL): {len(bear_obs)}')
````

## Step 6 — FVG (Fair Value Gap) Detection

**FVG** = شکاف ارزش منصفانه

FVG سه کندل می‌خواهد:
- کندل ۱ و کندل ۳ همپوشانی ندارند
- Gap = بین High[1] و Low[3] (برای Bullish FVG)

FVG نشان می‌دهد که قیمت با سرعت خارج شده و بعداً ممکن است برگردد.

````python
def detect_fvg(df: pd.DataFrame) -> pd.DataFrame:
    """
    Detect Fair Value Gaps.
    Bullish FVG:  high[i-2] < low[i]  (gap between candle 1 and 3)
    Bearish FVG:  low[i-2]  > high[i]
    Mark at bar i (the candle completing the gap).
    """
    df = df.copy()
    df['fvg_type']     = None
    df['fvg_top']      = np.nan
    df['fvg_bottom']   = np.nan
    df['fvg_midpoint'] = np.nan

    for i in range(2, len(df)):
        h_prev2 = df.iloc[i - 2]['high']
        l_prev2 = df.iloc[i - 2]['low']
        l_curr  = df.iloc[i]['low']
        h_curr  = df.iloc[i]['high']

        # Bullish FVG
        if h_prev2 < l_curr:
            df.at[i, 'fvg_type']     = 'bullish'
            df.at[i, 'fvg_top']      = l_curr
            df.at[i, 'fvg_bottom']   = h_prev2
            df.at[i, 'fvg_midpoint'] = (l_curr + h_prev2) / 2

        # Bearish FVG
        elif l_prev2 > h_curr:
            df.at[i, 'fvg_type']     = 'bearish'
            df.at[i, 'fvg_top']      = l_prev2
            df.at[i, 'fvg_bottom']   = h_curr
            df.at[i, 'fvg_midpoint'] = (l_prev2 + h_curr) / 2

    return df


df_m5 = detect_fvg(df_m5)

bull_fvg = (df_m5['fvg_type'] == 'bullish').sum()
bear_fvg = (df_m5['fvg_type'] == 'bearish').sum()
print(f'FVGs detected:')
print(f'  Bullish FVG: {bull_fvg}')
print(f'  Bearish FVG: {bear_fvg}')
print(f'  Avg FVG size: {df_m5["fvg_top"].dropna().mean() - df_m5["fvg_bottom"].dropna().mean():.2f}')
````

## Step 7 — Backtesting Engine

جریان کامل برای هر Order Block:
1. OB شناسایی می‌شود
2. منتظر Retest قیمت به OB هستیم
3. Rejection Candle در OB تأیید می‌شود
4. ورود، SL، TP محاسبه می‌شود

````python
def simulate_trade(df, entry_idx, direction, entry, sl, tp):
    for i in range(entry_idx, min(entry_idx + MAX_TRADE_BARS, len(df))):
        bar = df.iloc[i]
        if direction == 'BUY':
            if bar['low'] <= sl:
                return 'SL', -1.0, i - entry_idx + 1, i
            if bar['high'] >= tp:
                return 'TP', float(RISK_REWARD), i - entry_idx + 1, i
        else:
            if bar['high'] >= sl:
                return 'SL', -1.0, i - entry_idx + 1, i
            if bar['low'] <= tp:
                return 'TP', float(RISK_REWARD), i - entry_idx + 1, i
    last = min(entry_idx + MAX_TRADE_BARS, len(df)) - 1
    return 'OPEN', 0.0, MAX_TRADE_BARS, last


def is_rejection_candle(bar, ob_type, wick_ratio=REJECTION_WICK_RATIO):
    if bar['range'] == 0:
        return False
    if ob_type == 'bullish':
        return bar['lower_wick'] / bar['range'] > wick_ratio
    else:
        return bar['upper_wick'] / bar['range'] > wick_ratio


def run_ob_backtest(df, order_blocks):
    trades = []
    n = len(df)
    for ob in order_blocks:
        displaced = False
        for i in range(ob.ob_bar_idx + 1, min(ob.ob_bar_idx + OB_EXPIRY_BARS, n - 1)):
            bar = df.iloc[i]
            if ob.ob_type == 'bullish':
                if not displaced:
                    if bar['close'] > ob.ob_high:
                        displaced = True
                    continue
                if bar['low'] <= ob.ob_high and bar['close'] >= ob.ob_low:
                    if is_rejection_candle(bar, ob.ob_type):
                        entry = ob.ob_high
                        sl    = ob.ob_low - SL_BUFFER
                        risk  = entry - sl
                        tp    = entry + risk * RISK_REWARD
                        result, pnl_r, bars_held, exit_bar = simulate_trade(df, i + 1, 'BUY', entry, sl, tp)
                        fvg_conf = bool(
                            (df.iloc[max(0, ob.ob_bar_idx - 10):i + 1]['fvg_type'] == 'bullish').any()
                        )
                        trades.append({
                            'direction': 'BUY', 'entry_price': entry,
                            'sl': sl, 'tp': tp, 'result': result,
                            'pnl_r': pnl_r, 'bars_held': bars_held,
                            'entry_time': df.iloc[i + 1]['time'],
                            'exit_time':  df.iloc[exit_bar]['time'],
                            'retest_time': bar['time'], 'ob_time': ob.ob_time,
                            'ob_type': ob.ob_type, 'fvg_confluence': fvg_conf,
                        })
                        break
                if bar['close'] < ob.ob_low:
                    break
            else:
                if not displaced:
                    if bar['close'] < ob.ob_low:
                        displaced = True
                    continue
                if bar['high'] >= ob.ob_low and bar['close'] <= ob.ob_high:
                    if is_rejection_candle(bar, ob.ob_type):
                        entry = ob.ob_low
                        sl    = ob.ob_high + SL_BUFFER
                        risk  = sl - entry
                        tp    = entry - risk * RISK_REWARD
                        result, pnl_r, bars_held, exit_bar = simulate_trade(df, i + 1, 'SELL', entry, sl, tp)
                        fvg_conf = bool(
                            (df.iloc[max(0, ob.ob_bar_idx - 10):i + 1]['fvg_type'] == 'bearish').any()
                        )
                        trades.append({
                            'direction': 'SELL', 'entry_price': entry,
                            'sl': sl, 'tp': tp, 'result': result,
                            'pnl_r': pnl_r, 'bars_held': bars_held,
                            'entry_time': df.iloc[i + 1]['time'],
                            'exit_time':  df.iloc[exit_bar]['time'],
                            'retest_time': bar['time'], 'ob_time': ob.ob_time,
                            'ob_type': ob.ob_type, 'fvg_confluence': fvg_conf,
                        })
                        break
                if bar['close'] > ob.ob_high:
                    break
    return pd.DataFrame(trades)


trades_df = run_ob_backtest(df_m5, order_blocks)

if trades_df.empty:
    print('No trades generated.')
else:
    print(f'Total trades: {len(trades_df)}')
    print(f'  BUY  : {(trades_df["direction"]=="BUY").sum()}')
    print(f'  SELL : {(trades_df["direction"]=="SELL").sum()}')
    print(f'  TP   : {(trades_df["result"]=="TP").sum()}')
    print(f'  SL   : {(trades_df["result"]=="SL").sum()}')
    print(f'  FVG confluence: {trades_df["fvg_confluence"].sum()}')
    display(trades_df.head(5))
    print("...")
    display(trades_df.tail(5))
````

## Step 8 — Performance Analytics

````python
def calc_metrics(trades_df: pd.DataFrame) -> dict:
    if trades_df.empty:
        return {}
    closed = trades_df[trades_df['result'].isin(['TP', 'SL'])].copy()
    if closed.empty:
        return {'open_trades': (trades_df['result'] == 'OPEN').sum()}
    n    = len(closed)
    wins = (closed['result'] == 'TP').sum()
    wr   = wins / n
    closed['cum_r'] = closed['pnl_r'].cumsum()
    dd   = closed['cum_r'] - closed['cum_r'].cummax()
    pos  = closed[closed['pnl_r'] > 0]['pnl_r'].sum()
    neg  = abs(closed[closed['pnl_r'] < 0]['pnl_r'].sum())
    pf   = pos / neg if neg > 0 else float('inf')
    arr  = (closed['result'] == 'SL').astype(int).values
    max_cl = streak = 0
    for v in arr:
        streak = (streak + 1) if v else 0
        max_cl = max(max_cl, streak)
    return {
        'total_trades'   : n,
        'wins'           : int(wins),
        'losses'         : n - int(wins),
        'win_rate'       : wr,
        'total_r'        : round(closed['pnl_r'].sum(), 3),
        'avg_r'          : round(closed['pnl_r'].mean(), 3),
        'profit_factor'  : round(pf, 3),
        'max_dd_r'       : round(dd.min(), 3),
        'max_consec_loss': max_cl,
        'expectancy'     : round(wr * RISK_REWARD - (1 - wr), 3),
        'avg_bars'       : round(closed['bars_held'].mean(), 1),
        'cum_r'          : closed['cum_r'].reset_index(drop=True),
        'drawdown'       : dd.reset_index(drop=True),
        'closed'         : closed,
    }


metrics = calc_metrics(trades_df)

if metrics and metrics.get('total_trades', 0) > 0:
    sep = '=' * 55
    print(sep)
    print(f' {SYMBOL} - {LOOKBACK_DAYS}days - PERFORMANCE — Order Block Reaction Strategy')
    print(sep)
    print(f'  Trades        : {metrics["total_trades"]}')
    print(f'  Wins / Losses : {metrics["wins"]} / {metrics["losses"]}')
    print(f'  Win Rate      : {metrics["win_rate"]*100:.1f}%')
    print(f'  Total R       : {metrics["total_r"]:+.2f} R')
    print(f'  Avg R         : {metrics["avg_r"]:+.3f} R/trade')
    print(f'  Profit Factor : {metrics["profit_factor"]:.2f}')
    print(f'  Expectancy    : {metrics["expectancy"]:+.3f} R')
    print(f'  Max Drawdown  : {metrics["max_dd_r"]:.2f} R')
    print(f'  Avg Duration  : {metrics["avg_bars"]:.0f} M5 bars')
    if 'fvg_confluence' in trades_df.columns:
        fvg_sub = trades_df[trades_df['fvg_confluence'] & trades_df['result'].isin(['TP','SL'])]
        if len(fvg_sub) > 0:
            print(f'\n  FVG Confluence Filter:')
            print(f'    Trades with FVG: {len(fvg_sub)}')
            print(f'    WR with FVG    : {(fvg_sub["result"]=="TP").mean()*100:.1f}%')
    print(sep)
````

## Step 8.5 — Commission-Adjusted Performance (Errante Broker)

محاسبه سود/ضرر واقعی پس از کسر اسپرد و کمیسیون بروکر **Errante** — معامله با **0.02 لات**.

| Account | Spread | Commission | Min Deposit |
|---|---|---|---|
| Standard | 2.5 pips | $0 | €50 |
| Premium | 1.0 pip | $0 | €100 |
| VIP | 0.8 pips | $0 | €5,000 |
| Tailor Made | 0.0 pips | ~$7/lot RT | €15,000 |

> **XAUUSD pip convention used here**: 1 pip = $0.10 price move  
> → pip value for 0.02 lot = **$0.20 / pip** (contract size 100 oz × $0.10 × 0.02 lot)

````python
# ── Errante Commission Config ────────────────────────────────────────────────
LOT_SIZE      = 0.02
CONTRACT_SIZE = 100      # oz per standard lot
PIP_SIZE      = 0.10     # 1 pip = $0.10 price move for XAUUSD
PIP_VALUE     = LOT_SIZE * CONTRACT_SIZE * PIP_SIZE   # $/pip for chosen lot

INITIAL_CAPITAL = 5_000

# commission_rt = round-trip commission in $/lot
ERRANTE_ACCOUNTS = {
    'Standard':    {'spread_pips': 2.5, 'commission_rt': 0.0},
    'Premium':     {'spread_pips': 1.0, 'commission_rt': 0.0},
    'VIP':         {'spread_pips': 0.8, 'commission_rt': 0.0},
    'Tailor Made': {'spread_pips': 0.0, 'commission_rt': 7.0},
}


def commission_analysis(trades_df: pd.DataFrame, acc_name: str, acc: dict) -> pd.DataFrame:
    closed = trades_df[trades_df['result'].isin(['TP', 'SL'])].copy()
    if closed.empty:
        return pd.DataFrame()

    spread_cost = acc['spread_pips'] * PIP_VALUE   # fixed $ spread per trade
    comm_cost   = acc['commission_rt'] * LOT_SIZE  # fixed $ commission per trade
    total_cost  = spread_cost + comm_cost

    # Per-trade P&L breakdown
    closed['dollar_risk']    = abs(closed['entry_price'] - closed['sl']) * CONTRACT_SIZE * LOT_SIZE
    closed['pnl_usd_gross']  = closed['pnl_r'] * closed['dollar_risk']
    closed['spread_cost_usd']= spread_cost        # how much spread took per trade
    closed['commission_usd'] = comm_cost          # how much commission took per trade
    closed['total_cost_usd'] = total_cost
    closed['pnl_usd_net']    = closed['pnl_usd_gross'] - total_cost
    closed['cost_r']         = total_cost / closed['dollar_risk']
    closed['pnl_r_net']      = closed['pnl_r'] - closed['cost_r']
    closed['cum_usd_gross']  = closed['pnl_usd_gross'].cumsum()
    closed['cum_usd_net']    = closed['pnl_usd_net'].cumsum()
    closed['cum_r_net']      = closed['pnl_r_net'].cumsum()
    closed['account']        = acc_name
    closed['lot_size']       = LOT_SIZE
    return closed


account_results = {
    name: commission_analysis(trades_df, name, params)
    for name, params in ERRANTE_ACCOUNTS.items()
}

print(f'XAUUSD pip convention : 1 pip = ${PIP_SIZE}  →  pip value for {LOT_SIZE} lot = ${PIP_VALUE:.2f}/pip')
print(f'Spread formula        : spread_pips × pip_value = spread cost per trade')
print(f'Commission formula    : commission_rt ($/lot) × lot_size = commission per trade')
print(f'Capital for Return%   : ${INITIAL_CAPITAL:,}\n')

# ── Spread / Commission Cost Breakdown ───────────────────────────────────────
print('COST BREAKDOWN PER ACCOUNT (for 0.02 lot, per trade):')
print(f'  {"Account":<14} {"Spread pips":>12} {"Spread $":>10} {"Comm $/lot":>12} {"Comm $":>8} {"Total Cost":>11}')
print(f'  {"-" * 65}')
for acc_name, acc in ERRANTE_ACCOUNTS.items():
    sp = acc['spread_pips'] * PIP_VALUE
    co = acc['commission_rt'] * LOT_SIZE
    print(f'  {acc_name:<14} {acc["spread_pips"]:>10.1f}p {sp:>10.2f} '
          f'{acc["commission_rt"]:>10.1f}    {co:>8.2f} {sp+co:>11.2f}')

# ── Summary Table ─────────────────────────────────────────────────────────────
sep = '=' * 115
print(f'\n{sep}')
print(f'  ERRANTE — Performance Summary  |  Lot: {LOT_SIZE}  |  {SYMBOL}  |  Capital: ${INITIAL_CAPITAL:,}')
print(sep)
print(f'  {"Account":<14} {"Cost$/tr":>9} {"TotSprd$":>9} {"TotComm$":>9} '
      f'{"TotCost$":>9} {"Gross$":>9} {"Net$":>9} {"Return%":>9} {"WR%":>7} {"PF(net)":>8}')
print(f'  {"-" * 103}')

for acc_name, df_acc in account_results.items():
    if df_acc.empty:
        continue
    n        = len(df_acc)
    cost_tr  = df_acc['total_cost_usd'].iloc[0]
    tot_sprd = df_acc['spread_cost_usd'].iloc[0] * n
    tot_comm = df_acc['commission_usd'].iloc[0] * n
    tot_cost = cost_tr * n
    gross    = df_acc['pnl_usd_gross'].sum()
    net_d    = df_acc['pnl_usd_net'].sum()
    ret_pct  = net_d / INITIAL_CAPITAL * 100
    wr       = (df_acc['result'] == 'TP').mean() * 100
    pos      = df_acc[df_acc['pnl_usd_net'] > 0]['pnl_usd_net'].sum()
    neg      = abs(df_acc[df_acc['pnl_usd_net'] < 0]['pnl_usd_net'].sum())
    pf       = pos / neg if neg > 0 else float('inf')
    sign     = '+' if ret_pct >= 0 else ''
    print(f'  {acc_name:<14} {cost_tr:>9.2f} {tot_sprd:>9.2f} {tot_comm:>9.2f} '
          f'{tot_cost:>9.2f} {gross:>9.2f} {net_d:>9.2f} {sign}{ret_pct:>7.2f}% {wr:>7.1f}% {pf:>8.2f}')
print(sep)
print(f'  TotSprd$ = total spread paid across all {n} trades  |  TotComm$ = total commission paid')
print(f'  Gross$   = P&L before any costs         |  Net$     = Gross$ − TotCost$')

# ── Per-Trade Detail Table (all accounts side-by-side per trade) ──────────────
print(f'\n{"=" * 115}')
print(f'  PER-TRADE BREAKDOWN  |  Gross$ / Spread$ / Commission$ / Net$ per trade  |  Lot: {LOT_SIZE}')
print(f'{"=" * 115}')

# Build side-by-side header
accs = list(account_results.keys())
hdr  = f'  {"#":>3} {"Dir":>5} {"Entry":>9} {"Res":>5} {"Risk$":>7}'
for a in accs:
    short = a.replace('Tailor Made','TailorMd').replace('Standard','Std').replace('Premium','Prem')
    hdr  += f'  {short[:8]:>8}Gr$ {short[:8]:>8}Sp$ {short[:8]:>8}Co$ {short[:8]:>8}Net$'
print(hdr[:200])  # limit width

# Simpler: just show 4 Net$ columns side by side + Gross for first account
ref = list(account_results.values())[0]
print(f'\n  {"#":>3} {"Dir":>5} {"Entry":>9} {"Res":>5} {"Risk$":>7} '
      f'{"Gross$":>8} | '
      f'{"Std Net$":>9} {"Prem Net$":>10} {"VIP Net$":>9} {"TM Net$":>9} | '
      f'{"TM Sprd$":>9} {"TM Comm$":>9} {"CumNetTM$":>10}')
print(f'  {"-" * 108}')

tm = account_results['Tailor Made']
std = account_results['Standard']
prem = account_results['Premium']
vip = account_results['VIP']

for idx in range(len(tm)):
    row    = tm.iloc[idx]
    r_std  = std.iloc[idx]
    r_prem = prem.iloc[idx]
    r_vip  = vip.iloc[idx]
    gross  = row['pnl_usd_gross']
    g_sign = '+' if gross >= 0 else ''
    res_icon = 'TP' if row['result'] == 'TP' else 'SL'
    print(f'  {idx+1:>3} {row["direction"]:>5} {row["entry_price"]:>9.2f} {res_icon:>5} '
          f'{row["dollar_risk"]:>7.2f} '
          f'{g_sign}{gross:>7.2f} | '
          f'{r_std["pnl_usd_net"]:>+9.2f} {r_prem["pnl_usd_net"]:>+10.2f} '
          f'{r_vip["pnl_usd_net"]:>+9.2f} {row["pnl_usd_net"]:>+9.2f} | '
          f'{row["spread_cost_usd"]:>9.2f} {row["commission_usd"]:>9.2f} {row["cum_usd_net"]:>+10.2f}')

print(f'  {"-" * 108}')
total_gross = tm['pnl_usd_gross'].sum()
print(f'  {"TOTAL":>19} {"":>5} {"":>7} '
      f'{total_gross:>+8.2f} | '
      f'{std["pnl_usd_net"].sum():>+9.2f} {prem["pnl_usd_net"].sum():>+10.2f} '
      f'{vip["pnl_usd_net"].sum():>+9.2f} {tm["pnl_usd_net"].sum():>+9.2f} | '
      f'{tm["spread_cost_usd"].sum():>9.2f} {tm["commission_usd"].sum():>9.2f} '
      f'{tm["cum_usd_net"].iloc[-1]:>+10.2f}')
print(f'{"=" * 115}')
print(f'  Gross$ = P&L with no costs  |  Std/Prem/VIP/TM = net after that account\'s cost')
print(f'  TM Sprd$ = $0 (Tailor Made has 0 spread)  |  TM Comm$ = ${tm["commission_usd"].iloc[0]:.2f}/trade')
````

## Step 8.8 ? Single-Position Backtest (Live Bot Mode)

Simulates the live bot constraint: **only one open position at a time**, and picks the **most recent valid OB** per bar (matches `run_ob_xauusd.py` after BUG-1 fix).

Key differences vs the original multi-position backtest:
- Bar-by-bar scan instead of OB-by-OB
- `reversed(order_blocks)` ? most recent OB wins when multiple fire on the same bar
- New signal is blocked while a position is open (mirrors `position_open` guard in live bot)
- Each OB can only generate one trade (same as `seen_obs` in live bot)

````python
# ?? Single-Position Backtest: mirrors live bot behaviour ????????????????????
#
# Rules (identical to run_ob_xauusd.py after all bug-fixes):
#   1. Pick MOST RECENT valid OB per bar (reversed order_blocks)
#   2. Only ONE position open at a time ? new signals blocked while position is live
#   3. Each OB produces at most one trade (ob_traded flag = seen_obs equivalent)
#   4. SL_BUFFER = 0.0  (matches backtest & fixed live bot)
# ---------------------------------------------------------------------------

def run_ob_backtest_single_pos(df, obs):
    """Bar-by-bar backtest with single-position constraint (live bot mode)."""
    n = len(df)
    ob_displaced   = [False] * len(obs)
    ob_invalidated = [False] * len(obs)
    ob_traded      = [False] * len(obs)

    trades = []
    pos_exit_bar = -1          # bar index when the current position closes

    for bar_idx in range(1, n):
        bar = df.iloc[bar_idx]

        # ?? 1. Check for a new signal (only when no position is open) ??????
        if bar_idx > pos_exit_bar:
            for j in range(len(obs) - 1, -1, -1):   # reversed = most-recent OB first
                ob = obs[j]
                if not ob_displaced[j] or ob_invalidated[j] or ob_traded[j]:
                    continue
                if ob.ob_bar_idx >= bar_idx:
                    continue
                if bar_idx - ob.ob_bar_idx > OB_EXPIRY_BARS:
                    continue

                sig = None
                if ob.ob_type == "bullish":
                    if bar["low"] <= ob.ob_high and bar["close"] >= ob.ob_low:
                        if is_rejection_candle(bar, ob.ob_type):
                            entry = ob.ob_high
                            sl    = ob.ob_low - SL_BUFFER
                            risk  = entry - sl
                            tp    = entry + risk * RISK_REWARD
                            sig   = ("BUY", entry, sl, tp)
                else:
                    if bar["high"] >= ob.ob_low and bar["close"] <= ob.ob_high:
                        if is_rejection_candle(bar, ob.ob_type):
                            entry = ob.ob_low
                            sl    = ob.ob_high + SL_BUFFER
                            risk  = sl - entry
                            tp    = entry - risk * RISK_REWARD
                            sig   = ("SELL", entry, sl, tp)

                if sig:
                    direction, entry, sl, tp = sig
                    entry_bar = bar_idx + 1
                    if entry_bar >= n:
                        ob_traded[j] = True
                        break
                    result, pnl_r, bars_held, exit_bar = simulate_trade(
                        df, entry_bar, direction, entry, sl, tp
                    )
                    fvg_type = "bullish" if direction == "BUY" else "bearish"
                    fvg_conf = bool(
                        (df.iloc[max(0, ob.ob_bar_idx - 10):bar_idx + 1]["fvg_type"] == fvg_type).any()
                    )
                    trades.append({
                        "direction":   direction,
                        "entry_price": entry,
                        "sl":          sl,
                        "tp":          tp,
                        "result":      result,
                        "pnl_r":       pnl_r,
                        "bars_held":   bars_held,
                        "entry_time":  df.iloc[entry_bar]["time"],
                        "exit_time":   df.iloc[exit_bar]["time"],
                        "retest_time": bar["time"],
                        "ob_time":     ob.ob_time,
                        "ob_type":     ob.ob_type,
                        "fvg_confluence": fvg_conf,
                    })
                    pos_exit_bar = exit_bar
                    ob_traded[j] = True
                    break   # only one trade per bar

        # ?? 2. Update displacement / invalidation state ???????????????????
        for j, ob in enumerate(obs):
            if ob_invalidated[j] or ob_traded[j]:
                continue
            if ob.ob_bar_idx >= bar_idx:
                continue
            if bar_idx - ob.ob_bar_idx > OB_EXPIRY_BARS:
                continue
            if not ob_displaced[j]:
                if ob.ob_type == "bullish" and bar["close"] > ob.ob_high:
                    ob_displaced[j] = True
                elif ob.ob_type == "bearish" and bar["close"] < ob.ob_low:
                    ob_displaced[j] = True
            else:
                if ob.ob_type == "bullish" and bar["close"] < ob.ob_low:
                    ob_invalidated[j] = True
                elif ob.ob_type == "bearish" and bar["close"] > ob.ob_high:
                    ob_invalidated[j] = True

    return pd.DataFrame(trades)


# ?? Run ??????????????????????????????????????????????????????????????????????
trades_sp = run_ob_backtest_single_pos(df_m5, order_blocks)

print(f"Single-pos trades : {len(trades_sp)}")
if not trades_sp.empty:
    print(f"  BUY  : {(trades_sp['direction']=='BUY').sum()}")
    print(f"  SELL : {(trades_sp['direction']=='SELL').sum()}")
    print(f"  TP   : {(trades_sp['result']=='TP').sum()}")
    print(f"  SL   : {(trades_sp['result']=='SL').sum()}")


# ?? Metrics ??????????????????????????????????????????????????????????????????
metrics_sp = calc_metrics(trades_sp)

print()
if metrics_sp and metrics_sp.get("total_trades", 0) > 0:
    sep = "=" * 55
    print(sep)
    print(f"  SINGLE-POS MODE ? {SYMBOL} | {LOOKBACK_DAYS} days")
    print(sep)
    print(f"  Trades        : {metrics_sp['total_trades']}")
    print(f"  Wins / Losses : {metrics_sp['wins']} / {metrics_sp['losses']}")
    print(f"  Win Rate      : {metrics_sp['win_rate']*100:.1f}%")
    print(f"  Total R       : {metrics_sp['total_r']:+.2f} R")
    print(f"  Avg R         : {metrics_sp['avg_r']:+.3f} R/trade")
    print(f"  Profit Factor : {metrics_sp['profit_factor']:.2f}")
    print(f"  Expectancy    : {metrics_sp['expectancy']:+.3f} R")
    print(f"  Max Drawdown  : {metrics_sp['max_dd_r']:.2f} R")
    print(f"  Avg Duration  : {metrics_sp['avg_bars']:.0f} M5 bars")
    print(sep)


# Compute df_sp_tm early so Net$ is available for the comparison table
df_sp_tm = None
if not trades_sp.empty:
    acc_params = ERRANTE_ACCOUNTS["Tailor Made"]
    df_sp_tm   = commission_analysis(trades_sp, "Tailor Made", acc_params)

# ?? Side-by-side comparison ???????????????????????????????????????????????????
print()
print("=" * 60)
print("  COMPARISON: Multi-Position  vs  Single-Position (Live Bot)")
print("=" * 60)
m  = metrics       # original multi-pos
ms = metrics_sp    # single-pos

_mp_tm  = account_results.get("Tailor Made")
_mp_net = _mp_tm["pnl_usd_net"].sum() if _mp_tm is not None and not _mp_tm.empty else 0.0
_sp_net = df_sp_tm["pnl_usd_net"].sum() if df_sp_tm is not None and not df_sp_tm.empty else 0.0
_mp_ret = _mp_net / INITIAL_CAPITAL * 100
_sp_ret = _sp_net / INITIAL_CAPITAL * 100

rows = [
    ("Total Trades",  m.get("total_trades",0),            ms.get("total_trades",0)),
    ("Wins",          m.get("wins",0),                    ms.get("wins",0)),
    ("Losses",        m.get("losses",0),                  ms.get("losses",0)),
    ("Win Rate %",    f"{m.get('win_rate',0)*100:.1f}%",  f"{ms.get('win_rate',0)*100:.1f}%"),
    ("Total R",       f"{m.get('total_r',0):+.2f}",       f"{ms.get('total_r',0):+.2f}"),
    ("Avg R/trade",   f"{m.get('avg_r',0):+.3f}",         f"{ms.get('avg_r',0):+.3f}"),
    ("Profit Factor", f"{m.get('profit_factor',0):.2f}",  f"{ms.get('profit_factor',0):.2f}"),
    ("Expectancy",    f"{m.get('expectancy',0):+.3f}",    f"{ms.get('expectancy',0):+.3f}"),
    ("Max Drawdown R",f"{m.get('max_dd_r',0):.2f}",       f"{ms.get('max_dd_r',0):.2f}"),
    ("Avg Duration",  f"{m.get('avg_bars',0):.0f} bars",  f"{ms.get('avg_bars',0):.0f} bars"),
    ("Net$ (TM)",     f"{_mp_net:+.2f}",                   f"{_sp_net:+.2f}"),
    ("Return% (TM)",  f"{'+' if _mp_ret>=0 else ''}{_mp_ret:.2f}%", f"{'+' if _sp_ret>=0 else ''}{_sp_ret:.2f}%"),
]

print(f"  {'Metric':<20} {'Multi-Pos':>15} {'Single-Pos':>15}")
print(f"  {'-'*52}")
for label, v_multi, v_single in rows:
    print(f"  {label:<20} {str(v_multi):>15} {str(v_single):>15}")
print("=" * 60)


# ?? Save ?????????????????????????????????????????????????????????????????????
SP_DIR = OUT_BASE / "single_pos"
SP_DIR.mkdir(parents=True, exist_ok=True)

trades_sp.to_csv(SP_DIR / "trades_raw.csv", index=False)
print(f"\nSaved: {SP_DIR / 'trades_raw.csv'}  ({len(trades_sp)} rows)")

if df_sp_tm is not None:
    df_sp_tm.to_csv(SP_DIR / "trades_Tailor_Made.csv", index=False)
    print(f"Saved: {SP_DIR / 'trades_Tailor_Made.csv'}")

display(trades_sp)
````

## Step 8.9 — FVG Confluence Filter Backtest (4 Variants)

Runs **4 backtest variants** side by side:

| # | Mode | FVG Filter |
|---|------|------------|
| A | Multi-position (original) | No |
| B | Multi-position | Yes |
| C | Single-position (live bot) | No |
| D | Single-position (live bot) | Yes |

> **Multi-pos + FVG**: post-filtering `trades_df` is equivalent to a built-in filter
> because each OB is independent.
>
> **Single-pos + FVG**: must re-run bar-by-bar with the filter inside the loop —
> skipping a no-FVG signal frees the slot for the next valid OB.

````python
# ══════════════════════════════════════════════════════════════════════════
# Step 8.9 — FVG Confluence Filter Backtest (4 variants)
# ══════════════════════════════════════════════════════════════════════════

# ── Variant B: Multi-pos + FVG  (post-filter — equivalent to built-in) ───
trades_mp_fvg  = trades_df[trades_df['fvg_confluence']].reset_index(drop=True)
metrics_mp_fvg = calc_metrics(trades_mp_fvg)

print(f"Multi-pos ALL : {len(trades_df):>4} trades  |  "
      f"FVG: {trades_df['fvg_confluence'].sum():>3}  "
      f"no-FVG: {(~trades_df['fvg_confluence']).sum():>3}")


# ── Variant D: Single-pos + FVG  (must run full simulation) ─────────────
# Logic per bar (no open position):
#   valid retest + FVG    -> take trade, mark ob_traded, break
#   valid retest, no FVG  -> mark ob_traded (retest consumed), continue to next OB
#   no retest             -> continue to next OB

def run_ob_backtest_sp_fvg(df, obs):
    n = len(df)
    ob_displaced   = [False] * len(obs)
    ob_invalidated = [False] * len(obs)
    ob_traded      = [False] * len(obs)
    trades = []
    pos_exit_bar = -1

    for bar_idx in range(1, n):
        bar = df.iloc[bar_idx]

        # Signal check
        if bar_idx > pos_exit_bar:
            for j in range(len(obs) - 1, -1, -1):  # most-recent OB first
                ob = obs[j]
                if not ob_displaced[j] or ob_invalidated[j] or ob_traded[j]:
                    continue
                if ob.ob_bar_idx >= bar_idx:
                    continue
                if bar_idx - ob.ob_bar_idx > OB_EXPIRY_BARS:
                    continue

                retest = False
                if ob.ob_type == 'bullish':
                    if bar['low'] <= ob.ob_high and bar['close'] >= ob.ob_low:
                        retest = is_rejection_candle(bar, ob.ob_type)
                else:
                    if bar['high'] >= ob.ob_low and bar['close'] <= ob.ob_high:
                        retest = is_rejection_candle(bar, ob.ob_type)

                if not retest:
                    continue

                # Valid retest — check FVG
                fvg_type = 'bullish' if ob.ob_type == 'bullish' else 'bearish'
                fvg_conf = bool(
                    (df.iloc[max(0, ob.ob_bar_idx - 10):bar_idx + 1]['fvg_type'] == fvg_type).any()
                )
                ob_traded[j] = True  # retest consumed regardless of FVG result

                if not fvg_conf:
                    continue  # no FVG: try next OB for this bar

                # FVG present — build signal
                if ob.ob_type == 'bullish':
                    entry, sl, direction = ob.ob_high, ob.ob_low - SL_BUFFER, 'BUY'
                else:
                    entry, sl, direction = ob.ob_low, ob.ob_high + SL_BUFFER, 'SELL'
                risk = abs(entry - sl)
                tp = entry + risk * RISK_REWARD if direction == 'BUY' else entry - risk * RISK_REWARD

                entry_bar = bar_idx + 1
                if entry_bar >= n:
                    break

                result, pnl_r, bars_held, exit_bar = simulate_trade(
                    df, entry_bar, direction, entry, sl, tp
                )
                fvg_type_fvg = 'bullish' if direction == 'BUY' else 'bearish'
                trades.append({
                    'direction':      direction,
                    'entry_price':    entry,
                    'sl':             sl,
                    'tp':             tp,
                    'result':         result,
                    'pnl_r':          pnl_r,
                    'bars_held':      bars_held,
                    'entry_time':     df.iloc[entry_bar]['time'],
                    'exit_time':      df.iloc[exit_bar]['time'],
                    'retest_time':    bar['time'],
                    'ob_time':        ob.ob_time,
                    'ob_type':        ob.ob_type,
                    'fvg_confluence': True,
                })
                pos_exit_bar = exit_bar
                break  # one trade per bar

        # Update displacement / invalidation state
        for j, ob in enumerate(obs):
            if ob_invalidated[j] or ob_traded[j]:
                continue
            if ob.ob_bar_idx >= bar_idx:
                continue
            if bar_idx - ob.ob_bar_idx > OB_EXPIRY_BARS:
                continue
            if not ob_displaced[j]:
                if ob.ob_type == 'bullish' and bar['close'] > ob.ob_high:
                    ob_displaced[j] = True
                elif ob.ob_type == 'bearish' and bar['close'] < ob.ob_low:
                    ob_displaced[j] = True
            else:
                if ob.ob_type == 'bullish' and bar['close'] < ob.ob_low:
                    ob_invalidated[j] = True
                elif ob.ob_type == 'bearish' and bar['close'] > ob.ob_high:
                    ob_invalidated[j] = True

    return pd.DataFrame(trades)


trades_sp_fvg  = run_ob_backtest_sp_fvg(df_m5, order_blocks)
metrics_sp_fvg = calc_metrics(trades_sp_fvg)


# ── 4-variant comparison table ────────────────────────────────────────────
variants = [
    ('A  Multi-pos          ', trades_df,     metrics),
    ('B  Multi-pos + FVG    ', trades_mp_fvg, metrics_mp_fvg),
    ('C  Single-pos         ', trades_sp,     metrics_sp),
    ('D  Single-pos + FVG   ', trades_sp_fvg, metrics_sp_fvg),
]

print()
sep = '=' * 82
print(sep)
print(f"  4-VARIANT COMPARISON  |  {SYMBOL}  |  {LOOKBACK_DAYS} days")
print(sep)
print(f"  {'Variant':<26} {'Trades':>7} {'WR%':>7} {'Total R':>9} "
      f"{'Avg R':>8} {'PF':>7} {'Expect':>8} {'MaxDD':>8}")
print(f"  {'-'*77}")
for label, tdf, m in variants:
    if not m or m.get('total_trades', 0) == 0:
        print(f"  {label:<26} {'0':>7}  — no trades —")
        continue
    print(f"  {label:<26} {m['total_trades']:>7} "
          f"{m['win_rate']*100:>6.1f}% "
          f"{m['total_r']:>+9.2f} "
          f"{m['avg_r']:>+8.3f} "
          f"{m['profit_factor']:>7.2f} "
          f"{m['expectancy']:>+8.3f} "
          f"{m['max_dd_r']:>8.2f}")
print(sep)


# ── FVG filter impact ─────────────────────────────────────────────────────
print()
print('  FVG FILTER IMPACT:')
for base_label, fvg_label, m_base, m_fvg in [
    ('Multi-pos ', 'Multi-pos+FVG ', metrics,    metrics_mp_fvg),
    ('Single-pos', 'Single-pos+FVG', metrics_sp, metrics_sp_fvg),
]:
    if not m_base or not m_fvg or m_base.get('total_trades', 0) == 0:
        continue
    dropped  = m_base['total_trades'] - m_fvg.get('total_trades', 0)
    wr_delta = (m_fvg.get('win_rate', 0) - m_base['win_rate']) * 100
    r_delta  =  m_fvg.get('total_r',  0) - m_base['total_r']
    print(f"  {base_label}: -{dropped:>3} trades filtered  "
          f"WR {wr_delta:>+5.1f}%  Total R {r_delta:>+6.2f}")


# ── Save all 4 variants ───────────────────────────────────────────────────
FVG_DIR = OUT_BASE / 'fvg_filter'
FVG_DIR.mkdir(parents=True, exist_ok=True)

save_map = [
    ('A_multi_pos.csv',       trades_df),
    ('B_multi_pos_fvg.csv',   trades_mp_fvg),
    ('C_single_pos.csv',      trades_sp),
    ('D_single_pos_fvg.csv',  trades_sp_fvg),
]
print()
for fname, tdf in save_map:
    tdf.to_csv(FVG_DIR / fname, index=False)
    print(f'Saved: {FVG_DIR / fname}  ({len(tdf)} rows)')
print(f'\nAll 4 variants saved under: {FVG_DIR}')
````

## Step 8.6 — Capital & Leverage Simulation ($5,000 — 10x)

شبیه‌سازی سود/ضرر واقعی با **سرمایه $5,000** و **اهرم 10:1**

سه سناریو مقایسه می‌شود:

````python
# ── Capital & Leverage Parameters ────────────────────────────────────────────
INITIAL_CAPITAL = 5_000   # USD
LEVERAGE        = 10

gold_price = df_m5['close'].iloc[-1]

# ── Margin Calculation ────────────────────────────────────────────────────────
notional_per_lot  = gold_price * CONTRACT_SIZE          # $ value of 1 standard lot
margin_per_lot    = notional_per_lot / LEVERAGE         # margin needed for 1 lot
max_lot_by_margin = (INITIAL_CAPITAL * 0.70) / margin_per_lot   # keep 30% as buffer

# ── Risk-Based Lot Sizing ─────────────────────────────────────────────────────
closed_trades = trades_df[trades_df['result'].isin(['TP', 'SL'])].copy()
closed_trades['sl_dist'] = abs(closed_trades['entry_price'] - closed_trades['sl'])
avg_sl_dist = closed_trades['sl_dist'].mean()           # average SL in price units

# dollar_risk_per_lot = sl_distance × contract_size
dollar_risk_per_lot = avg_sl_dist * CONTRACT_SIZE

lot_1pct = (INITIAL_CAPITAL * 0.01) / dollar_risk_per_lot   # 1% risk lot
lot_2pct = (INITIAL_CAPITAL * 0.02) / dollar_risk_per_lot   # 2% risk lot

# round to nearest 0.01, cap at max allowed by margin
def snap(x): return max(0.01, round(round(x / 0.01) * 0.01, 2))

scenarios = {
    'Conservative (1% risk)': min(snap(lot_1pct), snap(max_lot_by_margin)),
    'Moderate     (2% risk)': min(snap(lot_2pct), snap(max_lot_by_margin)),
    'Max Margin   (70% cap)': snap(max_lot_by_margin),
}

# ── Print Margin Summary ──────────────────────────────────────────────────────
sep = '=' * 65
print(sep)
print(f'  MARGIN CALCULATOR  |  Capital: ${INITIAL_CAPITAL:,}  |  Leverage: {LEVERAGE}x')
print(sep)
print(f'  Gold price (last close) : ${gold_price:,.2f}')
print(f'  Notional value / 1 lot  : ${notional_per_lot:,.0f}  (1 lot = 100 oz)')
print(f'  Margin needed / 1 lot   : ${margin_per_lot:,.0f}')
print(f'  Avg SL distance (bt)    : {avg_sl_dist:.2f} pts → ${dollar_risk_per_lot:.0f}/lot risk')
print(f'  Lot for 1% risk ($50)   : {snap(lot_1pct):.2f}')
print(f'  Lot for 2% risk ($100)  : {snap(lot_2pct):.2f}')
print(f'  Max lot (70% margin)    : {snap(max_lot_by_margin):.2f}')
print(sep)

# ── Scaled P&L Table ─────────────────────────────────────────────────────────
print()
print('=' * 90)
print(f'  SCALED P&L  |  Capital ${INITIAL_CAPITAL:,} @ {LEVERAGE}x  |  Backtest window: {LOOKBACK_DAYS} days')
print('=' * 90)

for scen_name, lot in scenarios.items():
    scale       = lot / LOT_SIZE      # scale factor vs base 0.02 lot
    margin_used = margin_per_lot * lot
    margin_pct  = margin_used / INITIAL_CAPITAL * 100

    print(f'\n  ── {scen_name}  |  Lot: {lot:.2f}  |  '
          f'Margin: ${margin_used:,.0f} ({margin_pct:.0f}% of capital)  ──')
    print(f'  {"Account":<16} {"Cost$/tr":>9} {"NetUSD/5d":>11} '
          f'{"NetR":>7} {"MaxDD$":>9} {"Return%":>9} {"Ann%*":>8}')
    print(f'  {"-" * 73}')

    for acc_name, df_acc in account_results.items():
        if df_acc.empty:
            continue
        # scale all $ amounts linearly with lot size
        net_usd  = df_acc['pnl_usd_net'].sum() * scale
        net_r    = df_acc['pnl_r_net'].sum()
        max_dd   = (df_acc['cum_usd_net'] - df_acc['cum_usd_net'].cummax()).min() * scale
        cost_tr  = df_acc['total_cost_usd'].iloc[0] * scale
        ret_pct  = net_usd / INITIAL_CAPITAL * 100
        ann_pct  = ret_pct * (252 / 5)          # annualised (252 trading days)

        print(f'  {acc_name:<16} {cost_tr:>9.2f} {net_usd:>11.2f} '
              f'{net_r:>7.2f} {max_dd:>9.2f} {ret_pct:>8.2f}% {ann_pct:>7.0f}%')

print(f'\n{"=" * 90}')
print(f'  * Annualised = 5-day result × 252/5 trading days (theoretical, assumes same conditions)')
print(f'  * Return% = Net USD / ${INITIAL_CAPITAL:,} initial capital')
````

## Step 8.7 — Money Management Strategies Comparison

6 combination of position sizing on the same backtest trades — **Tailor Made** account (lowest cost).

| # | Strategy | Logic | Risk level |
|---|----------|-------|------------|
| 1 | **Fixed Base 0.02** | Always 0.02 lot — reference baseline | Low |
| 2 | **Fixed 1% Risk** | Lot = 1% of *initial* capital / trade dollar risk | Low |
| 3 | **Compound 1%** | Same but uses *current equity* — profits reinvested | Medium |
| 4 | **Half-Kelly** | Kelly% = WR − (1−WR)/RR, use half for safety | Medium |
| 5 | **Anti-Martingale** | After TP: lot × 1.5 (max 3 streak). After SL: reset | High |
| 6 | **Staged +0.02/$200** | Add 0.02 lot for every $200 net profit above start | Medium |

> **Kelly formula**: optimal risk% = WR − (1−WR)/RR

````python

# ── Step 8.7 — Money Management Simulation ───────────────────────────────────
import math

BASE_LOT        = 0.02
INITIAL_CAPITAL = 5_000
CONTRACT_SIZE   = 100
PIP_SIZE        = 0.10

# Use Tailor Made (lowest cost) as trade sequence
tm_trades = account_results['Tailor Made'].copy().reset_index(drop=True)

# Recover per-trade sl_dist from stored dollar_risk
# dollar_risk = sl_dist × CONTRACT_SIZE × lot → sl_dist = dollar_risk / (CONTRACT_SIZE × BASE_LOT)
tm_trades['sl_dist'] = tm_trades['dollar_risk'] / (CONTRACT_SIZE * BASE_LOT)

WIN_RATE  = (tm_trades['result'] == 'TP').mean()
RR        = 2.0
KELLY_PCT = max(0.0, WIN_RATE - (1 - WIN_RATE) / RR)
HALF_KELLY = KELLY_PCT / 2

def snap(x): return max(0.01, round(round(x / 0.01) * 0.01, 2))

# ── Strategy functions — each returns lot size for this trade ─────────────────

def strat_fixed_base(equity, base_lot, win_streak, lose_streak, trade_num):
    return BASE_LOT

def strat_fixed_1pct(equity, base_lot, win_streak, lose_streak, trade_num):
    dollar_risk = tm_trades.iloc[trade_num]['sl_dist'] * CONTRACT_SIZE
    return snap((INITIAL_CAPITAL * 0.01) / dollar_risk) if dollar_risk > 0 else BASE_LOT

def strat_compound_1pct(equity, base_lot, win_streak, lose_streak, trade_num):
    dollar_risk = tm_trades.iloc[trade_num]['sl_dist'] * CONTRACT_SIZE
    return snap((equity * 0.01) / dollar_risk) if dollar_risk > 0 else BASE_LOT

def strat_half_kelly(equity, base_lot, win_streak, lose_streak, trade_num):
    dollar_risk = tm_trades.iloc[trade_num]['sl_dist'] * CONTRACT_SIZE
    return snap((equity * HALF_KELLY) / dollar_risk) if dollar_risk > 0 else BASE_LOT

def strat_anti_martingale(equity, base_lot, win_streak, lose_streak, trade_num):
    if win_streak == 0:
        return BASE_LOT
    scale = min(1.5 ** min(win_streak, 3), 1.5 ** 3)
    return snap(BASE_LOT * scale)

def strat_staged(equity, base_lot, win_streak, lose_streak, trade_num):
    profit = equity - INITIAL_CAPITAL
    steps  = max(0, int(profit // 200))
    return snap(BASE_LOT + steps * 0.02)

STRATEGIES = {
    'Fixed Base 0.02'   : strat_fixed_base,
    'Fixed 1% Risk'     : strat_fixed_1pct,
    'Compound 1%'       : strat_compound_1pct,
    'Half-Kelly'        : strat_half_kelly,
    'Anti-Martingale'   : strat_anti_martingale,
    'Staged +0.02/$200' : strat_staged,
}

# ── Simulator ─────────────────────────────────────────────────────────────────

def simulate_mm(trades, strategy_fn, initial_capital=INITIAL_CAPITAL):
    equity       = initial_capital
    peak_equity  = initial_capital
    max_dd       = 0.0
    equity_curve = [initial_capital]
    win_streak   = 0
    lose_streak  = 0

    for i, row in trades.iterrows():
        lot       = strategy_fn(equity, BASE_LOT, win_streak, lose_streak, i)
        scale     = lot / BASE_LOT
        # net usd for this trade at the chosen lot size
        pnl_net   = row['pnl_usd_net'] * scale  # cost already inside pnl_usd_net

        equity   += pnl_net
        equity    = max(equity, 0.0)
        peak_equity = max(peak_equity, equity)
        dd = peak_equity - equity
        max_dd = max(max_dd, dd)

        if row['result'] == 'TP':
            win_streak  += 1
            lose_streak  = 0
        else:
            lose_streak += 1
            win_streak   = 0

        equity_curve.append(equity)

    return {
        'equity_curve': equity_curve,
        'final'       : equity,
        'net_pnl'     : equity - initial_capital,
        'return_pct'  : (equity - initial_capital) / initial_capital * 100,
        'max_dd_usd'  : max_dd,
        'max_dd_pct'  : max_dd / initial_capital * 100,
    }

results = {name: simulate_mm(tm_trades, fn) for name, fn in STRATEGIES.items()}

# ── Print Summary Table ────────────────────────────────────────────────────────
sep = '=' * 85
print(sep)
print(f'  MONEY MANAGEMENT COMPARISON  |  Capital: ${INITIAL_CAPITAL:,}  |  '
      f'Account: Tailor Made  |  Trades: {len(tm_trades)}')
print(f'  WR={WIN_RATE*100:.1f}%  RR={RR}  Kelly={KELLY_PCT*100:.2f}%  Half-Kelly={HALF_KELLY*100:.2f}%')
print(sep)
print(f'  {"Strategy":<22} {"Final $":>10} {"Net P&L":>10} {"Return%":>9} '
      f'{"MaxDD$":>9} {"MaxDD%":>8}')
print(f'  {"-" * 73}')

for name, r in results.items():
    sign = '+' if r['net_pnl'] >= 0 else ''
    print(f'  {name:<22} {r["final"]:>10,.2f} {sign}{r["net_pnl"]:>9,.2f} '
          f'{sign}{r["return_pct"]:>7.2f}% {r["max_dd_usd"]:>9,.2f} {r["max_dd_pct"]:>7.2f}%')

print(sep)
print(f'  Base lot: {BASE_LOT}  |  Half-Kelly lot sizing uses {HALF_KELLY*100:.2f}% equity per trade')

# ── Equity Curves — Plotly ─────────────────────────────────────────────────────
PALETTE = ['#00E5FF', '#00E676', '#FFAB40', '#7C4DFF', '#FF6D00', '#F50057']

fig = go.Figure()
for (name, r), color in zip(results.items(), PALETTE):
    fig.add_trace(go.Scatter(
        x=list(range(len(r['equity_curve']))),
        y=r['equity_curve'],
        mode='lines',
        name=f'{name}  (${r["final"]:,.0f})',
        line=dict(color=color, width=2),
    ))

fig.add_hline(y=INITIAL_CAPITAL, line_dash='dash', line_color='gray',
              annotation_text='Start $5,000', annotation_position='bottom right')

fig.update_layout(
    title=dict(
        text=(f'Money Management Strategy Comparison — Tailor Made Account<br>'
              f'<sup>Capital ${INITIAL_CAPITAL:,} | {len(tm_trades)} trades | '
              f'WR {WIN_RATE*100:.1f}% | RR {RR}</sup>'),
        x=0.5,
    ),
    xaxis_title='Trade Number',
    yaxis_title='Equity (USD)',
    template='plotly_dark',
    height=500,
    legend=dict(orientation='h', yanchor='bottom', y=1.02, x=0),
)
fig.show()
````

### 9.2 — Order Blocks + Displacements + FVGs on Chart

````python
def plot_ob_chart(df: pd.DataFrame, order_blocks: List[OrderBlock],
                  displacements: List[Displacement],
                  trades_df: pd.DataFrame,
                  tail_bars: int = 1000) -> None:
    """M5 chart with OBs, displacements, and FVGs visualized."""
    data     = df.tail(tail_bars).copy()
    min_t    = data['time'].min()
    max_t    = data['time'].max()

    fig = go.Figure()
    fig.add_trace(go.Candlestick(
        x=data['time'], open=data['open'], high=data['high'],
        low=data['low'], close=data['close'],
        name='M5',
        increasing_line_color='#26a69a',
        decreasing_line_color='#ef5350',
    ))

    # FVG zones
    fvg_data = data[data['fvg_type'].notna()]
    for _, fvg in fvg_data.iterrows():
        color = 'rgba(0,230,118,0.10)' if fvg['fvg_type'] == 'bullish' else 'rgba(255,23,68,0.10)'
        fig.add_shape(
            type='rect',
            x0=fvg['time'] - pd.Timedelta(minutes=5),
            x1=fvg['time'] + pd.Timedelta(minutes=25),
            y0=fvg['fvg_bottom'], y1=fvg['fvg_top'],
            fillcolor=color, line_width=0,
        )

    # Order Block zones
    for ob in order_blocks:
        if not (min_t <= pd.Timestamp(ob.ob_time) <= max_t):
            continue
        color  = 'rgba(0,176,255,0.20)'  if ob.ob_type == 'bullish' else 'rgba(239,83,80,0.20)'
        b_col  = 'rgba(0,176,255,0.6)'   if ob.ob_type == 'bullish' else 'rgba(239,83,80,0.6)'
        # OB box — extends forward until OB_EXPIRY_BARS
        end_t  = df.iloc[min(ob.ob_bar_idx + OB_EXPIRY_BARS, len(df)-1)]['time']
        fig.add_shape(
            type='rect',
            x0=ob.ob_time, x1=end_t,
            y0=ob.ob_low,  y1=ob.ob_high,
            fillcolor=color,
            line=dict(color=b_col, width=1),
        )
        label = 'Bull OB' if ob.ob_type == 'bullish' else 'Bear OB'
        fig.add_annotation(
            x=ob.ob_time, y=ob.ob_high if ob.ob_type == 'bearish' else ob.ob_low,
            text=label, showarrow=False,
            font=dict(size=8, color=b_col),
            bgcolor='rgba(0,0,0,0.5)',
        )

    # Displacement arrows
    for disp in displacements:
        if not (min_t <= pd.Timestamp(disp.start_time) <= max_t):
            continue
        color = '#26a69a' if disp.direction == 'UP' else '#ef5350'
        fig.add_annotation(
            ax=disp.start_time, ay=disp.start_price,
            x=disp.end_time,   y=disp.end_price,
            xref='x', yref='y', axref='x', ayref='y',
            showarrow=True, arrowhead=3, arrowsize=1.5,
            arrowwidth=2, arrowcolor=color,
        )

    # Trade entries
    if not trades_df.empty:
        for _, t in trades_df.iterrows():
            ec   = '#26a69a' if t['direction'] == 'BUY' else '#ef5350'
            esym = 'triangle-up' if t['direction'] == 'BUY' else 'triangle-down'
            rc   = '#00E676' if t['result'] == 'TP' else '#FF1744'
            fig.add_trace(go.Scatter(
                x=[t['retest_time']], y=[t['entry_price']],
                mode='markers', name='',
                marker=dict(symbol=esym, size=12, color=ec,
                            line=dict(color='white', width=1)),
                showlegend=False,
            ))

    fig.update_layout(
        title='Order Block Reaction — XAUUSD M5 (Blue=Bullish OB, Red=Bearish OB, Green=FVG)',
        xaxis_rangeslider_visible=False,
        template='plotly_dark', height=650,
        legend=dict(orientation='h', yanchor='bottom', y=1.01),
    )
    fig.show()


plot_ob_chart(df_m5, order_blocks, displacements, trades_df)
````

## Step 10 — FVG Confluence Analysis

````python
def plot_fvg_analysis(trades_df: pd.DataFrame) -> None:
    if trades_df.empty:
        return
    closed = trades_df[trades_df['result'].isin(['TP', 'SL'])].copy()
    if 'fvg_confluence' not in closed.columns or closed.empty:
        return

    fig = make_subplots(rows=1, cols=2,
                        subplot_titles=['Win Rate: OB only vs OB + FVG',
                                        'Avg R: OB only vs OB + FVG'])

    for col_i, metric_fn, label in [
        (1, lambda x: (x['result']=='TP').mean()*100, 'Win Rate %'),
        (2, lambda x: x['pnl_r'].mean(),              'Avg R'),
    ]:
        for fvg_flag, name, color in [
            (False, 'OB Only',   '#90a4ae'),
            (True,  'OB + FVG',  '#7C4DFF'),
        ]:
            sub = closed[closed['fvg_confluence'] == fvg_flag]
            val = metric_fn(sub) if len(sub) > 0 else 0
            fig.add_trace(go.Bar(
                x=[name], y=[val],
                text=[f'{val:.1f}' + ('%' if col_i==1 else 'R') + f' (n={len(sub)})'],
                textposition='auto',
                marker_color=color, name=name,
                showlegend=(col_i == 1),
            ), row=1, col=col_i)

    be = 100/(1+RISK_REWARD)
    fig.add_hline(y=be, line_dash='dash', line_color='yellow',
                  annotation_text=f'BE={be:.0f}%', row=1, col=1)
    fig.add_hline(y=0, line_dash='dash', line_color='gray', row=1, col=2)

    fig.update_layout(
        title='FVG Confluence Impact on Order Block Strategy',
        template='plotly_dark', height=400,
    )
    fig.show()


plot_fvg_analysis(trades_df)
````

# Temporal Performance Analysis

> **Source:** `results/08_order_block_reaction/XAUUSD/single_pos/trades_raw.csv`
>
> **Note:** All timestamps are in broker file-time. Per user specification,
> `19:00` in this dataset corresponds to **New York session open**.
>
> | Session   | File-time hours |
> |-----------|-----------------|
> | Asia      | 00:00 - 13:59   |
> | London    | 14:00 - 18:59   |
> | New York  | 19:00 - 23:59   |


````python
# ─────────────────────────────────────────────────────────────────────────────
# TEMPORAL ANALYSIS — Setup
# Loads trades_raw.csv, engineers features, defines session map and helpers.
# All new variables are prefixed _ta_ to avoid conflicts with existing cells.
# ─────────────────────────────────────────────────────────────────────────────
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.4f}'.format)

# ── Load ──────────────────────────────────────────────────────────────────────
_ta_csv = Path("results/08_order_block_reaction/XAUUSD/single_pos/trades_raw.csv")
_ta_raw = pd.read_csv(_ta_csv)
_ta_raw['entry_time'] = pd.to_datetime(_ta_raw['entry_time'], utc=True)

# ── Feature engineering ───────────────────────────────────────────────────────
_ta = _ta_raw.copy()
_ta['hour']       = _ta['entry_time'].dt.hour
_ta['dow']        = _ta['entry_time'].dt.dayofweek        # 0 = Monday
_ta['day_name']   = _ta['entry_time'].dt.day_name()
_ta['date']       = _ta['entry_time'].dt.date
_ta['is_win']     = (_ta['result'] == 'TP').astype(int)
_ta['pnl']        = _ta['pnl_r']
_ta['cum_pnl']    = _ta['pnl'].cumsum()

# ── Session classification ─────────────────────────────────────────────────────
# Broker file-time where NY Open = 19:00 (user specification)
#   Asia     00:00-13:59  (overnight + Tokyo)
#   London   14:00-18:59  (London open -> pre-NY)
#   New York 19:00-23:59  (NY open onwards)
def _ta_session(h):
    if h < 14:   return 'Asia'
    elif h < 19: return 'London'
    else:        return 'New York'

_ta['session'] = _ta['hour'].apply(_ta_session)

_TA_SESSION_ORDER  = ['Asia', 'London', 'New York']
_TA_SESSION_COLORS = {'Asia': '#388bfd', 'London': '#a5d6a7', 'New York': '#ef6c00'}

# ── 4-hour buckets ─────────────────────────────────────────────────────────────
_TA_BUCKET_DEFS  = [(0,3,'00-03'),(4,7,'04-07'),(8,11,'08-11'),
                    (12,15,'12-15'),(16,19,'16-19'),(20,23,'20-23')]
_TA_BUCKET_ORDER = [b[2] for b in _TA_BUCKET_DEFS]

def _ta_bucket(h):
    for lo, hi, lbl in _TA_BUCKET_DEFS:
        if lo <= h <= hi: return lbl

_ta['hour_bucket'] = _ta['hour'].apply(_ta_bucket)

_TA_DAY_ORDER = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']

# ── Summary helper ─────────────────────────────────────────────────────────────
def _ta_summarise(grp):
    n = len(grp)
    if n == 0:
        return pd.Series({'trades':0,'wins':0,'losses':0,'win_rate':np.nan,
                          'total_pnl':0.0,'avg_pnl':np.nan,'expectancy':np.nan})
    w  = int(grp['is_win'].sum())
    wr = w / n
    tp = grp['pnl'].sum()
    ap = grp['pnl'].mean()
    aw = grp.loc[grp['is_win']==1, 'pnl'].mean() if w > 0 else 0.0
    al = grp.loc[grp['is_win']==0, 'pnl'].mean() if (n - w) > 0 else 0.0
    ex = wr * aw + (1 - wr) * al
    return pd.Series({'trades':n, 'wins':w, 'losses':n-w,
                      'win_rate':round(wr,4), 'total_pnl':round(tp,2),
                      'avg_pnl':round(ap,4), 'expectancy':round(ex,4)})

print(f"Loaded {len(_ta)} trades  |  "
      f"{_ta['entry_time'].min().date()} -> {_ta['entry_time'].max().date()}")
print(f"Win rate : {_ta['is_win'].mean():.1%}   |   Total PnL : {_ta['pnl'].sum():.1f} R")
print()
print("Session distribution:")
print(_ta['session'].value_counts()[_TA_SESSION_ORDER].to_string())
````

````python
# ── 3. Session Analysis ───────────────────────────────────────────────────────
_ta_sess = (_ta.groupby('session', group_keys=False)
               .apply(_ta_summarise, include_groups=False)
               .reindex(_TA_SESSION_ORDER))
_ta_sess.index.name = 'Session'

print('SESSION PERFORMANCE')
print('=' * 70)
print(_ta_sess.rename(columns={
    'trades':'Trades','wins':'Wins','losses':'Losses',
    'win_rate':'Win Rate','total_pnl':'Total PnL (R)',
    'avg_pnl':'Avg PnL (R)','expectancy':'Expectancy'
}).to_string())

print()
best_sess  = _ta_sess['expectancy'].idxmax()
worst_sess = _ta_sess['expectancy'].idxmin()
print(f"  Best session  : {best_sess:<12}"
      f" [Exp {_ta_sess.loc[best_sess,'expectancy']:+.4f} R/trade]")
print(f"  Worst session : {worst_sess:<12}"
      f" [Exp {_ta_sess.loc[worst_sess,'expectancy']:+.4f} R/trade]")

# ── Max drawdown per session ──────────────────────────────────────────────────
def _ta_max_dd(pnl_series):
    cum = pnl_series.cumsum()
    dd  = (cum - cum.cummax())
    return round(dd.min(), 2)

print()
print('Max Drawdown per session (R):')
for s in _TA_SESSION_ORDER:
    dd = _ta_max_dd(_ta.loc[_ta['session']==s, 'pnl'])
    print(f"  {s:<12} : {dd:+.2f} R")

# ── Plot: Session overview ──────────────────────────────────────────────────
_sc = [_TA_SESSION_COLORS[s] for s in _TA_SESSION_ORDER]

fig = make_subplots(rows=2, cols=2,
    subplot_titles=['Total PnL (R)', 'Win Rate (%)',
                    'Trade Count', 'Expectancy (R/trade)'],
    vertical_spacing=0.16, horizontal_spacing=0.1)

fig.add_trace(go.Bar(x=_TA_SESSION_ORDER, y=_ta_sess['total_pnl'],
    marker_color=_sc,
    text=[f'{v:+.1f}' for v in _ta_sess['total_pnl']],
    textposition='outside', name='PnL'), row=1, col=1)
fig.add_hline(y=0, line_dash='dash', line_color='#8b949e', row=1, col=1)

fig.add_trace(go.Bar(x=_TA_SESSION_ORDER, y=_ta_sess['win_rate']*100,
    marker_color=_sc,
    text=[f'{v*100:.1f}%' for v in _ta_sess['win_rate']],
    textposition='outside', name='WR'), row=1, col=2)
fig.add_hline(y=50, line_dash='dash', line_color='#8b949e', row=1, col=2)

fig.add_trace(go.Bar(x=_TA_SESSION_ORDER, y=_ta_sess['trades'],
    marker_color=_sc,
    text=_ta_sess['trades'].astype(int).astype(str),
    textposition='outside', name='Trades'), row=2, col=1)

fig.add_trace(go.Bar(x=_TA_SESSION_ORDER, y=_ta_sess['expectancy'],
    marker_color=_sc,
    text=[f'{v:+.4f}' for v in _ta_sess['expectancy']],
    textposition='outside', name='Exp'), row=2, col=2)
fig.add_hline(y=0, line_dash='dash', line_color='#8b949e', row=2, col=2)

fig.update_layout(title='Session Performance Breakdown',
    height=620, showlegend=False,
    paper_bgcolor='#0d1117', plot_bgcolor='#161b22', font=dict(color='#c9d1d9'))
fig.update_yaxes(gridcolor='#21262d')
fig.show()

# ── Cumulative PnL by session ─────────────────────────────────────────────────
fig2 = go.Figure()
for s in _TA_SESSION_ORDER:
    sub = _ta[_ta['session']==s].copy().reset_index(drop=True)
    sub['cum'] = sub['pnl'].cumsum()
    fig2.add_trace(go.Scatter(x=sub.index, y=sub['cum'],
        mode='lines', name=s, line=dict(color=_TA_SESSION_COLORS[s], width=2)))

fig2.add_hline(y=0, line_dash='dash', line_color='#8b949e')
fig2.update_layout(title='Cumulative PnL by Session (trade sequence)',
    xaxis_title='Trade #', yaxis_title='Cumulative PnL (R)', height=420,
    paper_bgcolor='#0d1117', plot_bgcolor='#161b22', font=dict(color='#c9d1d9'),
    yaxis=dict(gridcolor='#21262d'), xaxis=dict(gridcolor='#21262d'))
fig2.show()

# ── PnL distribution boxplot ──────────────────────────────────────────────────
fig3 = go.Figure()
for s in _TA_SESSION_ORDER:
    fig3.add_trace(go.Box(y=_ta.loc[_ta['session']==s, 'pnl'],
        name=s, marker_color=_TA_SESSION_COLORS[s], boxmean='sd'))

fig3.update_layout(title='PnL Distribution by Session (Boxplot)',
    yaxis_title='PnL (R)', height=420,
    paper_bgcolor='#0d1117', plot_bgcolor='#161b22', font=dict(color='#c9d1d9'),
    yaxis=dict(gridcolor='#21262d'))
fig3.show()
````

````python
# ── 4. Open Session Behavior ─────────────────────────────────────────────────

_ta['london_open'] = _ta['hour'].between(14, 15)
_ta['ny_open']     = _ta['hour'].between(19, 20)

def _ta_open_label(row):
    if row['london_open']:         return 'London Open (14-15)'
    if row['ny_open']:             return 'NY Open (19-20)'
    if row['session'] == 'London': return 'London (rest 16-18)'
    if row['session'] == 'New York': return 'NY (rest 21-23)'
    return 'Asia'

_ta['open_label'] = _ta.apply(_ta_open_label, axis=1)

_OPEN_ORDER = ['London Open (14-15)', 'London (rest 16-18)',
               'NY Open (19-20)', 'NY (rest 21-23)', 'Asia']
_OPEN_COLORS = {
    'London Open (14-15)':   '#66bb6a',
    'London (rest 16-18)':   '#a5d6a7',
    'NY Open (19-20)':       '#ffa726',
    'NY (rest 21-23)':       '#ef6c00',
    'Asia':                  '#388bfd',
}

_ta_open_stats = (_ta.groupby('open_label', group_keys=False)
                     .apply(_ta_summarise, include_groups=False)
                     .reindex(_OPEN_ORDER)
                     .dropna(how='all'))

print('OPEN SESSION vs REST OF SESSION')
print('=' * 70)
print(_ta_open_stats.rename(columns={
    'trades':'Trades','wins':'Wins','losses':'Losses',
    'win_rate':'Win Rate','total_pnl':'Total PnL (R)',
    'avg_pnl':'Avg PnL (R)','expectancy':'Expectancy'
}).to_string())

# ── Open vs Rest comparison ────────────────────────────────────────────────────
print()
print('OPEN vs REST comparison:')
for open_lbl, rest_lbl in [('London Open (14-15)', 'London (rest 16-18)'),
                            ('NY Open (19-20)',     'NY (rest 21-23)')]:
    if open_lbl in _ta_open_stats.index and rest_lbl in _ta_open_stats.index:
        op  = _ta_open_stats.loc[open_lbl]
        rst = _ta_open_stats.loc[rest_lbl]
        sess = open_lbl.split(' Open')[0]
        print(f"  {sess} session:")
        print(f"    Open  -> WR {op['win_rate']*100:.1f}% | "
              f"Exp {op['expectancy']:+.4f} | Trades {int(op['trades'])}")
        print(f"    Rest  -> WR {rst['win_rate']*100:.1f}% | "
              f"Exp {rst['expectancy']:+.4f} | Trades {int(rst['trades'])}")
        verdict = 'Open is BETTER' if op['expectancy'] > rst['expectancy'] else 'Rest is BETTER'
        print(f"    Verdict: {verdict}")
        print()

# ── Plot: Open vs Rest ────────────────────────────────────────────────────────
_valid_labels = [l for l in _OPEN_ORDER if l in _ta_open_stats.index]
_oc = [_OPEN_COLORS[l] for l in _valid_labels]

fig = make_subplots(rows=1, cols=3,
    subplot_titles=['Expectancy (R/trade)', 'Win Rate (%)', 'Trade Count'],
    horizontal_spacing=0.1)

fig.add_trace(go.Bar(x=_valid_labels,
    y=_ta_open_stats.loc[_valid_labels, 'expectancy'],
    marker_color=_oc,
    text=[f'{v:+.4f}' for v in _ta_open_stats.loc[_valid_labels,'expectancy']],
    textposition='outside', name='Exp'), row=1, col=1)
fig.add_hline(y=0, line_dash='dash', line_color='#8b949e', row=1, col=1)

fig.add_trace(go.Bar(x=_valid_labels,
    y=_ta_open_stats.loc[_valid_labels, 'win_rate']*100,
    marker_color=_oc,
    text=[f'{v*100:.1f}%' for v in _ta_open_stats.loc[_valid_labels,'win_rate']],
    textposition='outside', name='WR'), row=1, col=2)
fig.add_hline(y=50, line_dash='dash', line_color='#8b949e', row=1, col=2)

fig.add_trace(go.Bar(x=_valid_labels,
    y=_ta_open_stats.loc[_valid_labels, 'trades'],
    marker_color=_oc,
    text=_ta_open_stats.loc[_valid_labels,'trades'].astype(int).astype(str),
    textposition='outside', name='Trades'), row=1, col=3)

fig.update_layout(title='Open Session vs Rest of Session',
    height=460, showlegend=False,
    paper_bgcolor='#0d1117', plot_bgcolor='#161b22', font=dict(color='#c9d1d9'))
fig.update_yaxes(gridcolor='#21262d')
fig.update_xaxes(tickangle=-15)
fig.show()

# ── Hourly detail: London-to-NY transition zone ───────────────────────────────
print('Hour-by-hour: London to NY transition zone (14-21):')
_oh = (_ta[_ta['hour'].between(14, 21)]
       .groupby('hour', group_keys=False)
       .apply(_ta_summarise, include_groups=False))
print(_oh[['trades','win_rate','avg_pnl','expectancy']].to_string())
````

## 6. Streak & Sequential Analysis

Checks whether losing/winning streaks cluster in specific sessions or hours.


````python
# ── 6. Streak & Sequential Analysis ──────────────────────────────────────────

_ta_seq = _ta[['entry_time','hour','session','day_name','is_win','pnl']].copy()
_ta_seq = _ta_seq.sort_values('entry_time').reset_index(drop=True)
_ta_seq['streak_type'] = _ta_seq['is_win'].map({1:'Win',0:'Loss'})

# ── Compute streak metadata ───────────────────────────────────────────────────
_ta_seq['streak_id']  = 0
_ta_seq['streak_len'] = 1

_sid, _idx = 0, 0
while _idx < len(_ta_seq):
    _cur = _ta_seq.iloc[_idx]['is_win']
    _j   = _idx
    while _j < len(_ta_seq) and _ta_seq.iloc[_j]['is_win'] == _cur:
        _j += 1
    _slen = _j - _idx
    for _k in range(_idx, _j):
        _ta_seq.at[_k, 'streak_id']  = _sid
        _ta_seq.at[_k, 'streak_len'] = _slen
    _sid += 1
    _idx  = _j

_streak_starts = _ta_seq.groupby('streak_id').first().reset_index()

# ── Text summary ──────────────────────────────────────────────────────────────
_loss_st = _streak_starts[_streak_starts['streak_type']=='Loss']
_win_st  = _streak_starts[_streak_starts['streak_type']=='Win']

print('STREAK SUMMARY')
print('=' * 60)
if len(_loss_st):
    print(f"  Losing streaks  : {len(_loss_st)}")
    print(f"    Max length    : {_loss_st['streak_len'].max()}")
    print(f"    Avg length    : {_loss_st['streak_len'].mean():.2f}")
if len(_win_st):
    print(f"  Winning streaks : {len(_win_st)}")
    print(f"    Max length    : {_win_st['streak_len'].max()}")
    print(f"    Avg length    : {_win_st['streak_len'].mean():.2f}")

print()
print('Losing streaks starting per session:')
print(_loss_st.groupby('session')['streak_len']
      .agg(['count','mean','max'])
      .reindex(_TA_SESSION_ORDER, fill_value=0)
      .to_string())

print()
print('Winning streaks starting per session:')
print(_win_st.groupby('session')['streak_len']
      .agg(['count','mean','max'])
      .reindex(_TA_SESSION_ORDER, fill_value=0)
      .to_string())

# ── Plot: Streak length distribution by session ───────────────────────────────
fig = make_subplots(rows=1, cols=2,
    subplot_titles=['Losing Streak Lengths by Session',
                    'Winning Streak Lengths by Session'],
    horizontal_spacing=0.12)

for s in _TA_SESSION_ORDER:
    fig.add_trace(go.Box(
        y=_loss_st[_loss_st['session']==s]['streak_len'],
        name=s, marker_color=_TA_SESSION_COLORS[s], showlegend=True),
        row=1, col=1)
    fig.add_trace(go.Box(
        y=_win_st[_win_st['session']==s]['streak_len'],
        name=s, marker_color=_TA_SESSION_COLORS[s], showlegend=False),
        row=1, col=2)

fig.update_layout(title='Streak Length Distribution by Session',
    height=440, paper_bgcolor='#0d1117', plot_bgcolor='#161b22',
    font=dict(color='#c9d1d9'),
    yaxis=dict(gridcolor='#21262d'), yaxis2=dict(gridcolor='#21262d'))
fig.show()

# ── Plot: All trades bar + cumulative line ─────────────────────────────────────
_colors_seq = ['#2ea043' if v == 1 else '#da3633' for v in _ta_seq['is_win']]
fig2 = go.Figure()
fig2.add_trace(go.Bar(x=_ta_seq.index, y=_ta_seq['pnl'],
    marker_color=_colors_seq, name='Trade PnL'))
fig2.add_trace(go.Scatter(x=_ta_seq.index, y=_ta_seq['pnl'].cumsum(),
    mode='lines', name='Cum PnL', line=dict(color='#f0e68c', width=2)))
fig2.add_hline(y=0, line_dash='dash', line_color='#8b949e')
fig2.update_layout(title='Sequential Trade PnL + Cumulative PnL',
    xaxis_title='Trade #', yaxis_title='R', height=420,
    paper_bgcolor='#0d1117', plot_bgcolor='#161b22', font=dict(color='#c9d1d9'),
    yaxis=dict(gridcolor='#21262d'), xaxis=dict(gridcolor='#21262d'))
fig2.show()
````

````python
# ── 8. Final Summary & Recommendations ───────────────────────────────────────

_TOTAL     = len(_ta)
_TOTAL_R   = _ta['pnl'].sum()
_GLOBAL_WR = _ta['is_win'].mean()
_GLOBAL_EXP = (
    _ta.loc[_ta['is_win']==1,'pnl'].mean() * _GLOBAL_WR +
    _ta.loc[_ta['is_win']==0,'pnl'].mean() * (1 - _GLOBAL_WR)
)
_days_span = (_ta['entry_time'].max() - _ta['entry_time'].min()).days

# Pull best/worst from earlier computed tables
_best_day   = _ta_dow['expectancy'].idxmax()
_worst_day  = _ta_dow['expectancy'].idxmin()
_best_sess  = _ta_sess['expectancy'].idxmax()
_worst_sess = _ta_sess['expectancy'].idxmin()
_best_bkt   = _ta_bucket_stats['expectancy'].idxmax()
_worst_bkt  = _ta_bucket_stats['expectancy'].idxmin()

sep = '=' * 72
sep2 = '-' * 72

print(sep)
print(' TEMPORAL ANALYSIS -- FINAL SUMMARY & RECOMMENDATIONS')
print(sep)
print()
print('DATASET OVERVIEW')
print(sep2)
print(f"  Period    : {_ta['entry_time'].min().date()} -> "
      f"{_ta['entry_time'].max().date()}  (~{_days_span} days)")
print(f"  Trades    : {_TOTAL}")
print(f"  Win Rate  : {_GLOBAL_WR*100:.1f}%")
print(f"  Total PnL : {_TOTAL_R:.1f} R")
print(f"  Expectancy: {_GLOBAL_EXP:+.4f} R/trade")

print()
print('KEY FINDINGS')
print(sep2)
print(f"  Best  day     : {_best_day}"
      f"  [Exp {_ta_dow.loc[_best_day,'expectancy']:+.4f} | "
      f"WR {_ta_dow.loc[_best_day,'win_rate']*100:.1f}%]")
print(f"  Worst day     : {_worst_day}"
      f"  [Exp {_ta_dow.loc[_worst_day,'expectancy']:+.4f} | "
      f"WR {_ta_dow.loc[_worst_day,'win_rate']*100:.1f}%]")
print(f"  Best  session : {_best_sess}"
      f"  [Exp {_ta_sess.loc[_best_sess,'expectancy']:+.4f} | "
      f"PnL {_ta_sess.loc[_best_sess,'total_pnl']:+.1f}R]")
print(f"  Worst session : {_worst_sess}"
      f"  [Exp {_ta_sess.loc[_worst_sess,'expectancy']:+.4f} | "
      f"PnL {_ta_sess.loc[_worst_sess,'total_pnl']:+.1f}R]")
print(f"  Best  bucket  : {_best_bkt}"
      f"  [Exp {_ta_bucket_stats.loc[_best_bkt,'expectancy']:+.4f}]")
print(f"  Worst bucket  : {_worst_bkt}"
      f"  [Exp {_ta_bucket_stats.loc[_worst_bkt,'expectancy']:+.4f}]")

# ── Filter simulations ────────────────────────────────────────────────────────
def _filter_sim(label, mask):
    kept    = _ta[mask]
    removed = _ta[~mask]
    nk = len(kept)
    wr_k = kept['is_win'].mean() if nk else float('nan')
    print(f"  Filter : {label}")
    print(f"    Kept    : {nk:>3} trades | PnL {kept['pnl'].sum():+.1f}R | "
          f"WR {wr_k*100:.1f}%")
    print(f"    Removed : {len(removed):>3} trades | "
          f"PnL {removed['pnl'].sum():+.1f}R")
    print()

print()
print('FILTER SIMULATIONS')
print(sep2)

# Worst session removed
_filter_sim(
    f"Remove '{_worst_sess}' session",
    _ta['session'] != _worst_sess)

# Worst day removed
_filter_sim(
    f"Remove '{_worst_day}'",
    _ta['day_name'] != _worst_day)

# Worst hour bucket removed
_wh_lo, _wh_hi = next(
    (lo, hi) for lo, hi, lbl in _TA_BUCKET_DEFS if lbl == _worst_bkt)
_filter_sim(
    f"Remove hour bucket '{_worst_bkt}' (hours {_wh_lo}-{_wh_hi})",
    ~_ta['hour'].isin(range(_wh_lo, _wh_hi+1)))

# Combined: worst session + worst day
_filter_sim(
    f"Remove '{_worst_sess}' AND '{_worst_day}'",
    (_ta['session'] != _worst_sess) & (_ta['day_name'] != _worst_day))

# ── Recommendations ───────────────────────────────────────────────────────────
print(sep)
print(' ACTIONABLE RECOMMENDATIONS')
print(sep)
print()
print('[1] PRIORITISE the best-performing session.')
print(f'    -> {_best_sess} showed the highest expectancy.')
print(f'    -> Consider reducing size or skipping {_worst_sess} sessions.')
print()
print('[2] DAY-OF-WEEK filter:')
print(f'    -> {_best_day} is the strongest day.')
print(f'    -> {_worst_day} is the weakest -- watch for lower conviction setups.')
print('    -> Friday afternoon near market close often has thin liquidity.')
print()
print('[3] HOUR BUCKET filter:')
print(f'    -> Worst performing bucket: {_worst_bkt}.')
print('    -> Consider skipping trades that enter in this window.')
print('    -> Validate with a backtest before applying live.')
print()
print('[4] OPEN SESSION caution:')
print('    -> London Open (14:00-15:59) and NY Open (19:00-20:59)')
print('       are high-volatility windows.')
print('    -> Compare open vs rest expectancy (see Section 4).')
print('    -> If open-hour WR < rest, consider waiting 1-2 candles.')
print()
print('[5] DATA VOLUME NOTE:')
print(f'    -> Dataset: ~{_days_span} days, {_TOTAL} trades.')
print('    -> Day-session combos with < 10 trades are statistically unreliable.')
print('    -> Focus on cells with >= 20 trades for actionable conclusions.')
print()
print('[6] NEXT STEPS:')
print('    -> Re-run this analysis monthly to detect regime changes.')
print('    -> Consider adding a news-event tag to separate fundamental')
print('       price moves from pure OB-reaction signals.')
print()
print(sep)
````

## 9. Monthly & Weekly Performance Breakdown

Categorises every closed trade (single-position mode) by **calendar month** and by the **week within that month** (W1 = days 1-7, W2 = 8-14, W3 = 15-21, W4 = 22+).

Metrics per bucket:
| Column | Description |
|---|---|
| **Trades** | Number of closed positions |
| **Win Rate** | TP% |
| **Net R** | Sum of pnl_r for the period |
| **Return %** | `(net USD PnL / $5 000 capital) × 100` |
| **Max DD (R)** | Peak-to-trough drawdown in R within the period |
| **Max DD %** | Same, expressed as % of initial capital |

````python
# ─────────────────────────────────────────────────────────────────────────────
# Step 9 — Monthly & Weekly Performance Breakdown (tables only)
# Source : single_pos / trades_Tailor_Made.csv
# ─────────────────────────────────────────────────────────────────────────────

import pandas as pd
import numpy as np
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

_MW_CAPITAL = 5_000   # USD

# ── Load ──────────────────────────────────────────────────────────────────────
_mw_path = Path("results/08_order_block_reaction/XAUUSD/single_pos/trades_Tailor_Made.csv")
_mw_raw  = pd.read_csv(_mw_path)
_mw_raw["entry_time"] = pd.to_datetime(_mw_raw["entry_time"], utc=True)
_mw = _mw_raw[_mw_raw["result"].isin(["TP", "SL"])].copy()

# ── Date features ─────────────────────────────────────────────────────────────
_mw["month_str"]     = _mw["entry_time"].dt.strftime("%Y-%m")
_mw["day_of_month"]  = _mw["entry_time"].dt.day
_mw["week_of_month"] = ((_mw["day_of_month"] - 1) // 7 + 1).clip(upper=4)
_mw["week_label"]    = "W" + _mw["week_of_month"].astype(str)

def _maxdd(series):
    if len(series) == 0:
        return 0.0
    cum  = np.cumsum(series.values)
    peak = np.maximum.accumulate(cum)
    return float(np.max(peak - cum))

def _metrics(grp):
    grp = grp.sort_values("entry_time")
    n   = len(grp)
    w   = (grp["result"] == "TP").sum()
    wr  = w / n if n else float("nan")
    net_r   = grp["pnl_r"].sum()
    net_usd = grp["pnl_usd_net"].sum()
    ret_pct = (net_usd / _MW_CAPITAL) * 100
    mdd_r   = _maxdd(grp["pnl_r"])
    mdd_pct = (_maxdd(grp["pnl_usd_net"]) / _MW_CAPITAL) * 100
    return pd.Series({
        "Trades"   : int(n),
        "W / L"    : f"{int(w)}/{int(n-w)}",
        "Win Rate" : f"{wr*100:.1f}%",
        "Net R"    : f"{net_r:+.2f}R",
        "Return %" : f"{ret_pct:+.2f}%",
        "Max DD R" : f"{-mdd_r:.2f}R",
        "Max DD %" : f"{-mdd_pct:.2f}%",
    })

sep  = "=" * 70
sep2 = "-" * 70

# ── Monthly summary ───────────────────────────────────────────────────────────
_monthly = (
    _mw.groupby("month_str", sort=True)
    .apply(_metrics, include_groups=False)
)
_monthly.index.name = "Month"

print(sep)
print("  MONTHLY SUMMARY  (Single-Position | Tailor Made | $5 000 capital)")
print(sep)
display(_monthly)
print()

# ── Weekly table (all months × weeks in one flat table) ───────────────────────
_weekly = (
    _mw.groupby(["month_str", "week_label"], sort=True)
    .apply(_metrics, include_groups=False)
)
_weekly.index.names = ["Month", "Week"]

print(sep)
print("  WEEKLY BREAKDOWN  (W1=days 1-7 | W2=8-14 | W3=15-21 | W4=22+)")
print(sep)
display(_weekly)
````

## 10. Slippage=3 — New TP Reachability Analysis

**Question:** After the new Tier-1 bot logic (slippage ≤ 3 pts → full lot, TP recalculated from fill), if every historically-winning trade had slippage = 3 pts, would all of them still reach the new (further) TP before their SL?

**New TP formula (from bot code):**
- BUY : fill = entry + 3  →  new_tp = fill + (fill − sl) × 2  →  shift = **+9 pts**
- SELL: fill = entry − 3  →  new_tp = fill − (sl − fill) × 2  →  shift = **−9 pts**

Scans M5 bars from each trade's entry_time and checks whether new_tp is hit before sl.

````python
# ─────────────────────────────────────────────────────────────────────────────
# 10. Slippage=3 — New TP Reachability Analysis
# Does every historically-winning trade still hit the new (9-pt-further) TP?
# ─────────────────────────────────────────────────────────────────────────────

import numpy as np
import pandas as pd
from pathlib import Path

_SLP_SLIP   = 3.0   # pts — slippage assumption
_SLP_RR     = 2.0   # risk-reward ratio (matches RISK_REWARD in bot)

# ── Load M5 OHLCV once ───────────────────────────────────────────────────────
_slp_m5 = pd.read_csv(DATA_DIR / SYMBOL / 'M5' / 'ohlcv.csv')
_slp_m5['time'] = pd.to_datetime(_slp_m5['time'], utc=True)
_slp_m5 = _slp_m5.sort_values('time').set_index('time')

# ── Helper: scan bars from entry, return True if new_tp hit before sl ────────
def _slp_tp_reached(direction, entry_time, sl, new_tp, m5):
    future = m5.loc[entry_time:]
    for bar in future.itertuples():
        if direction == 'BUY':
            if bar.low  <= sl:     return False   # SL first (conservative)
            if bar.high >= new_tp: return True
        else:
            if bar.high >= sl:     return False
            if bar.low  <= new_tp: return True
    return False  # ran out of data

# ── Compute new TP for both CSV sources ──────────────────────────────────────
_slp_sources = {
    'multi_pos': RESULT_DIR / '08_order_block_reaction' / SYMBOL / 'trades_raw.csv',
    'single_pos': RESULT_DIR / '08_order_block_reaction' / SYMBOL / 'single_pos' / 'trades_raw.csv',
}

for _slp_label, _slp_path in _slp_sources.items():
    df = pd.read_csv(_slp_path)
    df['entry_time'] = pd.to_datetime(df['entry_time'], utc=True)

    # compute fill price and new TP
    df['fill']         = np.where(df['direction'] == 'BUY',
                                    df['entry_price'] + _SLP_SLIP,
                                    df['entry_price'] - _SLP_SLIP)
    df['risk_from_fill'] = (df['fill'] - df['sl']).abs()
    df['new_tp']       = np.where(df['direction'] == 'BUY',
                                    df['fill'] + df['risk_from_fill'] * _SLP_RR,
                                    df['fill'] - df['risk_from_fill'] * _SLP_RR)
    df['tp_shift']     = (df['new_tp'] - df['tp']).round(2)

    wins = df[df['result'] == 'TP'].copy()
    wins['new_tp_reached'] = wins.apply(
        lambda r: _slp_tp_reached(r['direction'], r['entry_time'],
                                   r['sl'], r['new_tp'], _slp_m5),
        axis=1,
    )

    reached     = wins['new_tp_reached'].sum()
    not_reached = (~wins['new_tp_reached']).sum()
    total       = len(wins)

    print(f'══════════════════════════════════════════════')
    print(f'  Source      : {_slp_label}')
    print(f'  Winning trades (original): {total}')
    print(f'  TP shift BUY : +{wins.loc[wins["direction"]=="BUY","tp_shift"].mean():.1f} pts  |  ', end='')
    print(f'TP shift SELL: {wins.loc[wins["direction"]=="SELL","tp_shift"].mean():.1f} pts')
    print(f'  New TP reached (still win) : {reached:>3} / {total}  ({100*reached/total:.1f}%)')
    print(f'  New TP NOT reached (→ loss): {not_reached:>3} / {total}  ({100*not_reached/total:.1f}%)')

    for _d in ['BUY', 'SELL']:
        sub = wins[wins['direction'] == _d]
        if sub.empty: continue
        r = sub['new_tp_reached'].sum()
        print(f'    {_d:4s}: {r}/{len(sub)} ({100*r/len(sub):.1f}%)')

    miss = wins[~wins['new_tp_reached']]
    if miss.empty:
        print(f'✓ All {total} winning trades still reach new TP with slippage={_SLP_SLIP}.')
    else:
        print(f'✗ {len(miss)} trade(s) would NOT reach new TP:')
        print(miss[['direction','entry_time','entry_price','sl','tp','new_tp','tp_shift']].to_string(index=False))
````

## 11. Limit Order Fill-Rate Analysis

**Question:** If we replace the current market order entry with a **limit order** at the
original signal `entry_price`, what percentage of trades would actually be filled?

**Motivation:** The live bot receives 0-4 pts of slippage per market order. Placing a limit
order at the ideal OB-touch price avoids slippage but requires price to retrace back to that
level after the signal fires.

**Method:** For every trade in `trades_raw.csv`, M5 OHLCV bars are scanned from `entry_time`
over a 24-hour window (288 bars):
- **BUY limit**: `low <= entry_price` on any bar => order fills
- **SELL limit**: `high >= entry_price` on any bar => order fills

Trades that never touch `entry_price` within 24 hours are counted as **missed**.

````python
# ─────────────────────────────────────────────────────────────────────────────
# 11. Limit Order Fill-Rate Analysis
# ─────────────────────────────────────────────────────────────────────────────

import pandas as pd
import numpy as np
from pathlib import Path

# ── Config ────────────────────────────────────────────────────────────────────
_LO_WINDOW_BARS = 288    # 24-hour limit-order validity window (288 M5 bars)

# ── Load M5 OHLCV ─────────────────────────────────────────────────────────────
_lo_m5 = pd.read_csv(DATA_DIR / SYMBOL / "M5" / "ohlcv.csv")
_lo_m5["time"] = pd.to_datetime(_lo_m5["time"], utc=True)
_lo_m5 = _lo_m5.sort_values("time").set_index("time")
_lo_times = _lo_m5.index

# ── Load trades ───────────────────────────────────────────────────────────────
_lo_trades = pd.read_csv(RESULT_DIR / "08_order_block_reaction" / SYMBOL / "trades_raw.csv")
for _col in ["entry_time", "exit_time", "retest_time"]:
    _lo_trades[_col] = pd.to_datetime(_lo_trades[_col], utc=True)

# ── Scan each trade: would a limit order at entry_price have filled? ──────────
_lo_rows = []
for _, _row in _lo_trades.iterrows():
    _et  = _row["entry_time"]
    _ep  = _row["entry_price"]
    _dir = _row["direction"]

    _start = _lo_times.searchsorted(_et)
    _end   = min(_start + _LO_WINDOW_BARS, len(_lo_m5))
    _fut   = _lo_m5.iloc[_start:_end]

    _filled      = False
    _fill_bar    = None
    _max_adverse = 0.0   # pts price moved away before retracing to fill level

    for _i, (_t, _bar) in enumerate(_fut.iterrows()):
        if _dir == "BUY":
            if _bar.high > _ep:
                _max_adverse = max(_max_adverse, _bar.high - _ep)
            if _bar.low <= _ep:
                _filled, _fill_bar = True, _i
                break
        else:  # SELL
            if _bar.low < _ep:
                _max_adverse = max(_max_adverse, _ep - _bar.low)
            if _bar.high >= _ep:
                _filled, _fill_bar = True, _i
                break

    _lo_rows.append({
        **_row.to_dict(),
        "filled":          _filled,
        "bars_to_fill":    _fill_bar,
        "max_adverse_pts": _max_adverse if _filled else None,
    })

_lo        = pd.DataFrame(_lo_rows)
_lo_filled = _lo[_lo.filled].copy()
_lo_missed = _lo[~_lo.filled].copy()

_sep  = "=" * 62
_sep2 = "-" * 62

# ── 1. Overall Fill Rate ──────────────────────────────────────────────────────
print(_sep)
print("  LIMIT ORDER FILL-RATE ANALYSIS  (24-hour validity window)")
print(_sep)
print(f"  Total trades            : {len(_lo):>4}")
print(f"  Limit orders FILLED     : {_lo.filled.sum():>4}  ({_lo.filled.mean()*100:.1f}%)")
print(f"  Limit orders MISSED     : {(~_lo.filled).sum():>4}  ({(~_lo.filled).mean()*100:.1f}%)")
print()

# ── 2. By Direction ───────────────────────────────────────────────────────────
print(_sep2)
print("  Fill Rate by Direction")
print(_sep2)
_dir_rows = []
for _d in ["BUY", "SELL"]:
    _sub = _lo[_lo.direction == _d]
    _fs  = _sub[_sub.filled]
    _ms  = _sub[~_sub.filled]
    _dir_rows.append({
        "Direction":         _d,
        "Trades":            len(_sub),
        "Filled":            int(_fs.filled.sum()),
        "Fill %":            f"{_fs.filled.sum()/len(_sub)*100:.1f}%",
        "Missed":            len(_ms),
        "Missed all TP?":    "Yes" if len(_ms) > 0 and (_ms.result == "TP").all() else "No",
        "Avg bars to fill":  f"{_fs.bars_to_fill.mean():.1f}" if not _fs.empty else "—",
        "Avg adverse (pts)": f"{_fs.max_adverse_pts.mean():.2f}" if not _fs.empty else "—",
        "Med adverse (pts)": f"{_fs.max_adverse_pts.median():.2f}" if not _fs.empty else "—",
    })
display(pd.DataFrame(_dir_rows).set_index("Direction"))
print()

# ── 3. Time to Fill Distribution ─────────────────────────────────────────────
print(_sep2)
print("  Time to Fill  (M5 bars after signal | 1 bar = 5 min)")
print(_sep2)
_bins   = [-1, 0, 2, 5, 11, 23, 47, 287]
_labels = [
    "Same bar (0 bars)",
    "1-2 bars  (5-10 min)",
    "3-5 bars  (15-25 min)",
    "6-11 bars (30-55 min)",
    "12-23 bars (1-2 h)",
    "24-47 bars (2-4 h)",
    "48-287 bars (4-24 h)",
]
_lo_filled["fill_bucket"] = pd.cut(_lo_filled.bars_to_fill, bins=_bins, labels=_labels)
_fill_dist = (
    _lo_filled.fill_bucket
    .value_counts()
    .reindex(_labels)
    .rename_axis("Time window")
    .rename("Count")
    .to_frame()
)
_fill_dist["% of Filled"]  = (_fill_dist["Count"] / len(_lo_filled) * 100).round(1).astype(str) + "%"
_fill_dist["Cumulative %"] = (_fill_dist["Count"].cumsum() / len(_lo_filled) * 100).round(1).astype(str) + "%"
display(_fill_dist)
print()

# ── 4. Adverse Move Before Fill ──────────────────────────────────────────────
print(_sep2)
print("  Max Adverse Move BEFORE Fill (pts price moved away before retracing)")
print(_sep2)
_adv_rows = []
for _d in ["BUY", "SELL"]:
    _fs = _lo_filled[_lo_filled.direction == _d]
    for _thr in [0, 1, 2, 3, 5, 10]:
        _adv_rows.append({
            "Direction":       _d,
            "Adverse > N pts": _thr,
            "% of fills":      f"{(_fs.max_adverse_pts > _thr).mean()*100:.0f}%",
        })
display(
    pd.DataFrame(_adv_rows)
    .pivot(index="Adverse > N pts", columns="Direction", values="% of fills")
)
print()

# ── 5. Missed Trades Detail ───────────────────────────────────────────────────
print(_sep2)
print("  Missed Trades  (limit never reached within 24 h)")
print(_sep2)
if len(_lo_missed) == 0:
    print("  None — all trades would have filled.")
else:
    display(
        _lo_missed[["direction","entry_time","entry_price","sl","tp",
                     "result","pnl_r","bars_held"]]
        .reset_index(drop=True)
    )
print()

# ── 6. Strategy Performance: Market vs Limit ─────────────────────────────────
print(_sep2)
print("  Strategy Impact: Market Orders vs Limit Orders")
print(_sep2)
_all_tp = (_lo.result == "TP").sum()
_all_sl = (_lo.result == "SL").sum()
_all_wr = _all_tp / (_all_tp + _all_sl) * 100
_all_r  = _lo.pnl_r.sum()

_lim_tp = (_lo_filled.result == "TP").sum()
_lim_sl = (_lo_filled.result == "SL").sum()
_lim_wr = _lim_tp / (_lim_tp + _lim_sl) * 100
_lim_r  = _lo_filled.pnl_r.sum()

_missed_r = _lo_missed.pnl_r.sum()
_n_missed = len(_lo_missed)

_cmp = pd.DataFrame({
    "Metric": ["Trades", "Win Rate", "Total R", "Missed R"],
    "Market Orders (backtest)": [
        str(len(_lo)),
        f"{_all_wr:.1f}%",
        f"{_all_r:.1f}R",
        "—",
    ],
    "Limit Orders (simulated)": [
        f"{len(_lo_filled)} (-{_n_missed} missed)",
        f"{_lim_wr:.1f}%",
        f"{_lim_r:.1f}R",
        f"-{_missed_r:.1f}R ({_n_missed} TP trades)",
    ],
}).set_index("Metric")
display(_cmp)
print()

# ── 7. Summary & Recommendations ─────────────────────────────────────────────
print(_sep)
print("  FINDINGS & RECOMMENDATIONS")
print(_sep)
_fill_pct      = _lo.filled.mean() * 100
_buy_fill_pct  = _lo[_lo.direction == "BUY"].filled.mean() * 100
_sell_fill_pct = _lo[_lo.direction == "SELL"].filled.mean() * 100
_same_bar_pct  = (_lo_filled.bars_to_fill == 0).mean() * 100
_med_adv       = _lo_filled.max_adverse_pts.median()
_avg_bars      = _lo_filled.bars_to_fill.mean()

print(f"  Fill rate       : {_fill_pct:.1f}%  (BUY {_buy_fill_pct:.1f}%  |  SELL {_sell_fill_pct:.1f}%)")
print(f"  Same-bar fills  : {_same_bar_pct:.0f}%  (fill on the very first M5 bar after signal)")
print(f"  Avg bars fill   : {_avg_bars:.1f} bars  (~{_avg_bars*5/60:.1f} h after signal)")
print(f"  Median adverse  : {_med_adv:.1f} pts  (price moved away before retracing to fill)")
print(f"  Missed trades   : {_n_missed}  (ALL are TP / winning trades — no SL trades missed)")
print(f"  R lost to misses: -{_missed_r:.1f}R  of {_all_r:.1f}R total  ({_missed_r/_all_r*100:.1f}% of gross R)")
print()
print("KEY INSIGHT:")
print("  Every missed trade is a WINNER (TP). Limit orders only miss the cleanest,")
print("  fastest wins where price ran straight to TP without retracing.")
print("  This creates negative selection bias: limit orders skip winners,")
print("  while capturing all SL trades unchanged.")
print()
print("RECOMMENDATION:")
print("  1. Market orders remain the better execution choice for this strategy.")
print("  2. If live slippage > 3 pts is common, use a limit with a small buffer:")
print("       BUY : limit at entry_price - 1 pt  (just below OB level)")
print("       SELL: limit at entry_price + 1 pt  (just above OB level)")
print(f"     This should recover most of the {_n_missed} missed trades while")
print("     still avoiding worst-case slippage.")
print(f"  3. The {_same_bar_pct:.0f}% same-bar fill rate confirms price almost always")
print("     re-touches the OB level immediately after signal — a 0-1 pt")
print("     offset is sufficient to avoid slippage without missing trades.")
print(f"  4. Monitor the {_n_missed} missed trades in live trading. If they share")
print("     a pattern (gap opens, news spikes), add conditional logic:")
print("     fall back to market order if gap from entry_price > 3 pts.")
````

## 12. EMA-9 / DEMA-9 Trend Filter Analysis

**Question:** If we add a **trend filter** using EMA 9 or DEMA 9 on H1 or D1,
only taking signals aligned with the higher-timeframe trend, how does performance change?

**Filter Rule (no lookahead):**
- **BUY signal** → only trade if H1/D1 close **above** EMA/DEMA at signal time (uptrend)
- **SELL signal** → only trade if H1/D1 close **below** EMA/DEMA at signal time (downtrend)

**DEMA** (Double EMA) = 2 × EMA(n) − EMA(EMA(n)) — lower lag than a plain EMA.

**4 variants tested:**
| # | Indicator | Timeframe |
|---|-----------|----------|
| 1 | EMA 9    | H1       |
| 2 | DEMA 9   | H1       |
| 3 | EMA 9    | D1       |
| 4 | DEMA 9   | D1       |

Metrics compared: **Trades**, **Win Rate**, **Total R**, **Max DD (R)**, **Expectancy**, **Return% (at 1% risk/trade)**.

````python
# ─────────────────────────────────────────────────────────────────────────────
# 12. EMA-14 / DEMA-14 Trend Filter Analysis
# ─────────────────────────────────────────────────────────────────────────────

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

_EF_PERIOD = 9
_EF_RISK   = 1.0   # assumed % risk per trade (for Return% column)

# ── EMA / DEMA helpers ───────────────────────────────────────────────────────
def _ef_ema(s, n):
    return s.ewm(span=n, adjust=False).mean()

def _ef_dema(s, n):
    e1 = _ef_ema(s, n)
    return 2 * e1 - _ef_ema(e1, n)

# ── Load H1 and D1 data with EMA / DEMA ─────────────────────────────────────
_ef_ohlcv = {}
for _tf, _delta in [("H1", pd.Timedelta(hours=1)), ("D1", pd.Timedelta(days=1))]:
    _p = Path("data") / "XAUUSD" / _tf / "ohlcv.csv"
    _d = pd.read_csv(_p)
    _d["time"] = pd.to_datetime(_d["time"], utc=True)
    _d = _d.sort_values("time").reset_index(drop=True)
    _d["ema9"]      = _ef_ema(_d["close"], _EF_PERIOD)
    _d["dema9"]     = _ef_dema(_d["close"], _EF_PERIOD)
    # close_time = open_time + 1 period; merge on close_time avoids lookahead
    _d["close_time"] = _d["time"] + _delta
    _ef_ohlcv[_tf]   = _d[["close_time", "close", "ema9", "dema9"]].copy()
    print(f"{_tf}: {len(_d):,} bars  ({_d["time"].min().date()} → {_d["time"].max().date()})")

# ── Load trades ──────────────────────────────────────────────────────────────
_ef_csv    = Path("results/08_order_block_reaction/XAUUSD/single_pos/trades_raw.csv")
_ef_trades = pd.read_csv(_ef_csv)
_ef_trades["entry_time"] = pd.to_datetime(_ef_trades["entry_time"], utc=True)
_ef_trades = _ef_trades[_ef_trades["result"].isin(["TP", "SL"])].reset_index(drop=True)
print(f"Trades loaded: {len(_ef_trades)}")
print()

# ── Attach last-closed H1 / D1 EMA & DEMA values at each entry_time ─────────
for _tf in ["H1", "D1"]:
    _tf_df = _ef_ohlcv[_tf].rename(columns={"close_time": "entry_time"})
    _merged = pd.merge_asof(
        _ef_trades[["entry_time"]].sort_values("entry_time").reset_index(),
        _tf_df,
        on="entry_time",
        direction="backward"
    ).sort_values("index").reset_index(drop=True)
    _ef_trades[f"close_{_tf}"]  = _merged["close"].values
    _ef_trades[f"ema9_{_tf}"]  = _merged["ema9"].values
    _ef_trades[f"dema9_{_tf}"] = _merged["dema9"].values

# ── Metrics helper ───────────────────────────────────────────────────────────
def _ef_metrics(df, label, n_total):
    n    = len(df)
    if n == 0:
        return {"Label": label, "Trades": 0, "Filtered Out": n_total,
                "Win Rate": "—", "Total R": "—", "Max DD (R)": "—",
                "Expectancy": "—", "Return% (1% risk)": "—"}
    wins = (df["result"] == "TP").sum()
    wr   = wins / n
    r    = df["pnl_r"].sum()
    cum  = df["pnl_r"].cumsum()
    dd   = (cum - cum.cummax()).min()
    ex   = wr * 2 - (1 - wr)   # expectancy with RR=2
    rtn  = r * _EF_RISK         # Return% assuming _EF_RISK% risk per trade
    return {
        "Label":            label,
        "Trades":           n,
        "Filtered Out":     n_total - n,
        "Win Rate":         f"{wr*100:.1f}%",
        "Total R":          f"{r:+.1f}R",
        "Max DD (R)":       f"{dd:.1f}R",
        "Expectancy":       f"{ex:+.3f}",
        "Return% (1% risk)":f"{rtn:+.1f}%",
    }

# ── Build comparison table ────────────────────────────────────────────────────
_ef_n = len(_ef_trades)
_ef_rows = [_ef_metrics(_ef_trades, "Baseline (no filter)", _ef_n)]

for _tf in ["H1", "D1"]:
    for _ind_name, _ind in [("EMA", "ema9"), ("DEMA", "dema9")]:
        _col   = f"{_ind}_{_tf}"
        _label = f"{_ind_name} {_EF_PERIOD} — {_tf}"
        _mask  = (
            ((_ef_trades["direction"] == "BUY")  & (_ef_trades[f"close_{_tf}"] > _ef_trades[_col])) |
            ((_ef_trades["direction"] == "SELL") & (_ef_trades[f"close_{_tf}"] < _ef_trades[_col]))
        )
        _ef_rows.append(_ef_metrics(_ef_trades[_mask].reset_index(drop=True), _label, _ef_n))

_ef_result = pd.DataFrame(_ef_rows).set_index("Label")

# ── Print summary table ───────────────────────────────────────────────────────
_sep  = "=" * 72
_sep2 = "-" * 72
print(_sep)
print(f"  EMA-{_EF_PERIOD} / DEMA-{_EF_PERIOD} TREND FILTER — Performance Comparison")
print(_sep)
print("  Filter: BUY only when H1/D1 close > EMA/DEMA")
print("          SELL only when H1/D1 close < EMA/DEMA")
print("  Lookahead-free: uses last *fully closed* H1/D1 bar at entry time")
print()
display(_ef_result)

# ── Per-variant detail ────────────────────────────────────────────────────────
print()
print(_sep2)
print("  Direction breakdown for each variant")
print(_sep2)
_ef_dir_rows = []
for _tf in ["H1", "D1"]:
    for _ind_name, _ind in [("EMA", "ema9"), ("DEMA", "dema9")]:
        _col   = f"{_ind}_{_tf}"
        _label = f"{_ind_name} {_EF_PERIOD} — {_tf}"
        _mask  = (
            ((_ef_trades["direction"] == "BUY")  & (_ef_trades[f"close_{_tf}"] > _ef_trades[_col])) |
            ((_ef_trades["direction"] == "SELL") & (_ef_trades[f"close_{_tf}"] < _ef_trades[_col]))
        )
        for _d in ["BUY", "SELL"]:
            _sub = _ef_trades[_mask & (_ef_trades["direction"] == _d)]
            _base_sub = _ef_trades[_ef_trades["direction"] == _d]
            _n   = len(_sub)
            if _n > 0:
                _wr  = (_sub["result"] == "TP").sum() / _n
                _r   = _sub["pnl_r"].sum()
            else:
                _wr, _r = 0.0, 0.0
            _ef_dir_rows.append({
                "Filter":    _label,
                "Direction": _d,
                "Trades":    _n,
                "Kept %":    f"{_n/len(_base_sub)*100:.0f}%",
                "Win Rate":  f"{_wr*100:.1f}%",
                "Total R":   f"{_r:+.1f}R",
            })
_ef_dir_df = pd.DataFrame(_ef_dir_rows)
display(_ef_dir_df.set_index(["Filter", "Direction"]))

# ── Equity curves (all 5 variants on one chart) ───────────────────────────────
print()
print(_sep2)
print("  Equity Curves — Cumulative R")
print(_sep2)

_ef_fig = go.Figure()
_ef_colors = [
    ("#888888", "dash"),
    ("#2196F3", "solid"),
    ("#00BCD4", "dot"),
    ("#FF9800", "solid"),
    ("#F44336", "dot"),
]
_ef_variants = [
    ("Baseline (no filter)", _ef_trades),
]
for _tf in ["H1", "D1"]:
    for _ind_name, _ind in [("EMA", "ema9"), ("DEMA", "dema9")]:
        _col  = f"{_ind}_{_tf}"
        _mask = (
            ((_ef_trades["direction"] == "BUY")  & (_ef_trades[f"close_{_tf}"] > _ef_trades[_col])) |
            ((_ef_trades["direction"] == "SELL") & (_ef_trades[f"close_{_tf}"] < _ef_trades[_col]))
        )
        _ef_variants.append((f"{_ind_name} {_EF_PERIOD} — {_tf}", _ef_trades[_mask].reset_index(drop=True)))

for (_lbl, _df), (_clr, _dash) in zip(_ef_variants, _ef_colors):
    if len(_df) == 0:
        continue
    _cum = _df["pnl_r"].cumsum()
    _ef_fig.add_trace(go.Scatter(
        y=_cum, mode="lines", name=_lbl,
        line=dict(color=_clr, width=2, dash=_dash),
    ))

_ef_fig.add_hline(y=0, line_color="white", line_width=0.5, opacity=0.3)
_ef_fig.update_layout(
    title="EMA/DEMA Trend Filter — Equity Curves (Cumulative R)",
    xaxis_title="Trade #",
    yaxis_title="Cumulative R",
    template="plotly_dark",
    legend=dict(x=0.01, y=0.99),
    height=450,
)
_ef_fig.show()

# ── Best variant summary ──────────────────────────────────────────────────────
print()
print(_sep)
print("  FINDINGS")
print(_sep)
_ef_numeric = _ef_result.copy()
_ef_numeric["_r_val"] = _ef_numeric["Total R"].str.replace("R","").str.replace("+","").astype(float)
_ef_numeric["_dd_val"] = _ef_numeric["Max DD (R)"].str.replace("R","").astype(float)
_ef_numeric["_wr_val"] = _ef_numeric["Win Rate"].str.replace("%","").replace("—","0").astype(float)
_best_r  = _ef_numeric["_r_val"].idxmax()
_best_wr = _ef_numeric["_wr_val"].idxmax()
_best_dd = _ef_numeric["_dd_val"].idxmax()   # least negative = closest to 0
print(f"  Best Total R    : {_best_r}  → {_ef_result.loc[_best_r, 'Total R']}")
print(f"  Best Win Rate   : {_best_wr}  → {_ef_result.loc[_best_wr, 'Win Rate']}")
print(f"  Smallest Max DD : {_best_dd}  → {_ef_result.loc[_best_dd, 'Max DD (R)']}")
print()
print("  Baseline reference:")
print(f"    Trades  : {_ef_result.loc['Baseline (no filter)', 'Trades']}")
print(f"    Win Rate: {_ef_result.loc['Baseline (no filter)', 'Win Rate']}")
print(f"    Total R : {_ef_result.loc['Baseline (no filter)', 'Total R']}")
print(f"    Max DD  : {_ef_result.loc['Baseline (no filter)', 'Max DD (R)']}")
````

## 13. Advanced Filter Ideas — Win Rate, Return & Max DD Improvement

Tests **7 additional filters** beyond the EMA/DEMA trend filter, comparing win rate,
total R, max DD, and expectancy vs the unfiltered baseline.

| # | Filter | Logic |
|---|--------|-------|
| A | **FVG Confluence** | Only trade when an FVG overlaps the OB (already a column) |
| B | **RSI(14) H1** | BUY when H1 RSI > 50 (bullish momentum), SELL when RSI < 50 |
| C | **EMA(14) Slope D1** | D1 EMA must be *rising* for BUY and *falling* for SELL |
| D | **ATR Gate H1** | Only trade when H1 ATR(14) ≥ its rolling median (enough volatility) |
| E | **Session: London + Asia only** | Drop NY session (hour ≥ 19) — worst performer |
| F | **Combo: D1 EMA Slope + RSI H1** | Trend (slope) AND momentum aligned |
| G | **Combo: D1 EMA Slope + Session** | Trend (slope) AND no NY session |

All higher-TF lookups are **lookahead-free** (last fully-closed bar before entry).

````python
# ─────────────────────────────────────────────────────────────────────────────
# 13. Advanced Filter Ideas
# ─────────────────────────────────────────────────────────────────────────────

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

_AF_PERIOD   = 14
_AF_RISK_PCT = 1.0   # % account risked per trade (for Return% column)

# ─────────────────────────────────────────────────────────────────────────────
# Indicator helpers
# ─────────────────────────────────────────────────────────────────────────────
def _af_ema(s, n):
    return s.ewm(span=n, adjust=False).mean()

def _af_rsi(s, n=14):
    delta = s.diff()
    gain  = delta.clip(lower=0)
    loss  = (-delta).clip(lower=0)
    avg_g = gain.ewm(com=n - 1, adjust=False).mean()
    avg_l = loss.ewm(com=n - 1, adjust=False).mean()
    rs    = avg_g / avg_l.replace(0, np.nan)
    return 100 - (100 / (1 + rs))

def _af_atr(df, n=14):
    h, l, c = df["high"], df["low"], df["close"]
    pc = c.shift(1)
    tr = pd.concat([(h - l), (h - pc).abs(), (l - pc).abs()], axis=1).max(axis=1)
    return tr.ewm(span=n, adjust=False).mean()

# ─────────────────────────────────────────────────────────────────────────────
# Load H1 and D1 with all indicators
# ─────────────────────────────────────────────────────────────────────────────
_af_h1 = pd.read_csv(Path("data/XAUUSD/H1/ohlcv.csv"))
_af_h1["time"] = pd.to_datetime(_af_h1["time"], utc=True)
_af_h1 = _af_h1.sort_values("time").reset_index(drop=True)
_af_h1["ema14"]      = _af_ema(_af_h1["close"], _AF_PERIOD)
_af_h1["rsi14"]      = _af_rsi(_af_h1["close"], _AF_PERIOD)
_af_h1["atr14"]      = _af_atr(_af_h1,          _AF_PERIOD)
_af_h1["atr_med"]    = _af_h1["atr14"].rolling(200, min_periods=20).median()
_af_h1["close_time"] = _af_h1["time"] + pd.Timedelta(hours=1)

_af_d1 = pd.read_csv(Path("data/XAUUSD/D1/ohlcv.csv"))
_af_d1["time"] = pd.to_datetime(_af_d1["time"], utc=True)
_af_d1 = _af_d1.sort_values("time").reset_index(drop=True)
_af_d1["ema14"]      = _af_ema(_af_d1["close"], _AF_PERIOD)
_af_d1["ema14_prev"] = _af_d1["ema14"].shift(1)   # previous bar EMA (for slope)
_af_d1["close_time"] = _af_d1["time"] + pd.Timedelta(days=1)

print(f"H1: {len(_af_h1):,} bars | D1: {len(_af_d1):,} bars")

# ─────────────────────────────────────────────────────────────────────────────
# Load trades
# ─────────────────────────────────────────────────────────────────────────────
_af_csv    = Path("results/08_order_block_reaction/XAUUSD/single_pos/trades_raw.csv")
_af_trades = pd.read_csv(_af_csv)
_af_trades["entry_time"] = pd.to_datetime(_af_trades["entry_time"], utc=True)
_af_trades = _af_trades[_af_trades["result"].isin(["TP", "SL"])].reset_index(drop=True)
_af_n      = len(_af_trades)

# session label (broker time)
def _af_sess(h):
    if h < 14:   return "Asia"
    elif h < 19: return "London"
    return "NY"
_af_trades["hour"]    = _af_trades["entry_time"].dt.hour
_af_trades["session"] = _af_trades["hour"].apply(_af_sess)

# ─────────────────────────────────────────────────────────────────────────────
# Merge H1 indicators (lookahead-free: use close_time <= entry_time)
# ─────────────────────────────────────────────────────────────────────────────
_af_h1_merge = _af_h1[["close_time","close","ema14","rsi14","atr14","atr_med"]].rename(
    columns={"close_time": "entry_time"})
_af_merged_h1 = pd.merge_asof(
    _af_trades[["entry_time"]].sort_values("entry_time").reset_index(),
    _af_h1_merge,
    on="entry_time", direction="backward"
).sort_values("index").reset_index(drop=True)
_af_trades["h1_close"]   = _af_merged_h1["close"].values
_af_trades["h1_ema14"]   = _af_merged_h1["ema14"].values
_af_trades["h1_rsi14"]   = _af_merged_h1["rsi14"].values
_af_trades["h1_atr14"]   = _af_merged_h1["atr14"].values
_af_trades["h1_atr_med"] = _af_merged_h1["atr_med"].values

# ─────────────────────────────────────────────────────────────────────────────
# Merge D1 indicators (lookahead-free: use close_time <= entry_time)
# ─────────────────────────────────────────────────────────────────────────────
_af_d1_merge = _af_d1[["close_time","close","ema14","ema14_prev"]].rename(
    columns={"close_time": "entry_time"})
_af_merged_d1 = pd.merge_asof(
    _af_trades[["entry_time"]].sort_values("entry_time").reset_index(),
    _af_d1_merge,
    on="entry_time", direction="backward"
).sort_values("index").reset_index(drop=True)
_af_trades["d1_close"]     = _af_merged_d1["close"].values
_af_trades["d1_ema14"]     = _af_merged_d1["ema14"].values
_af_trades["d1_ema14_prev"]= _af_merged_d1["ema14_prev"].values

print(f"Trades: {_af_n} | Indicators merged.")
print()

# ─────────────────────────────────────────────────────────────────────────────
# Metrics helper
# ─────────────────────────────────────────────────────────────────────────────
def _af_metrics(df, label):
    n = len(df)
    if n == 0:
        return {"Filter": label, "Trades": 0, "Removed": _af_n,
                "Win Rate": "—", "Total R": "—", "Max DD": "—",
                "Expectancy": "—", "Return%": "—"}
    wins = (df["result"] == "TP").sum()
    wr   = wins / n
    r    = df["pnl_r"].sum()
    cum  = df["pnl_r"].cumsum()
    dd   = (cum - cum.cummax()).min()
    ex   = wr * 2 - (1 - wr)
    return {
        "Filter":     label,
        "Trades":     n,
        "Removed":    _af_n - n,
        "Win Rate":   f"{wr*100:.1f}%",
        "Total R":    f"{r:+.1f}R",
        "Max DD":     f"{dd:.1f}R",
        "Expectancy": f"{ex:+.3f}",
        "Return%":    f"{r * _AF_RISK_PCT:+.1f}%",
    }

# ─────────────────────────────────────────────────────────────────────────────
# Define all filter masks
# ─────────────────────────────────────────────────────────────────────────────
_t = _af_trades  # shorthand

# A — FVG confluence (column already exists)
_mA = _t["fvg_confluence"] == True

# B — RSI(14) H1 momentum alignment
_mB = (
    ((_t["direction"] == "BUY")  & (_t["h1_rsi14"] > 50)) |
    ((_t["direction"] == "SELL") & (_t["h1_rsi14"] < 50))
)

# C — D1 EMA(14) slope in trade direction (EMA must be rising/falling)
_mC = (
    ((_t["direction"] == "BUY")  & (_t["d1_ema14"] > _t["d1_ema14_prev"]) & (_t["d1_close"] > _t["d1_ema14"])) |
    ((_t["direction"] == "SELL") & (_t["d1_ema14"] < _t["d1_ema14_prev"]) & (_t["d1_close"] < _t["d1_ema14"]))
)

# D — ATR gate H1: only trade when volatility >= rolling median
_mD = _t["h1_atr14"] >= _t["h1_atr_med"]

# E — Session: Asia + London only (drop NY = hour >= 19)
_mE = _t["session"].isin(["Asia", "London"])

# F — Combo: D1 EMA slope + H1 RSI alignment
_mF = _mC & _mB

# G — Combo: D1 EMA slope + no NY
_mG = _mC & _mE

_af_filters = [
    ("Baseline (no filter)",          slice(None)),
    ("A — FVG Confluence",            _mA),
    ("B — RSI(14) H1 > / < 50",      _mB),
    ("C — D1 EMA Slope aligned",      _mC),
    ("D — H1 ATR >= median",          _mD),
    ("E — Session: Asia + London",    _mE),
    ("F — Combo: EMA Slope + RSI",    _mF),
    ("G — Combo: EMA Slope + Session",_mG),
]

# ─────────────────────────────────────────────────────────────────────────────
# Build summary table
# ─────────────────────────────────────────────────────────────────────────────
_af_rows = []
for _lbl, _mask in _af_filters:
    _sub = _t[_mask].reset_index(drop=True) if not isinstance(_mask, slice) else _t.copy()
    _af_rows.append(_af_metrics(_sub, _lbl))

_af_result = pd.DataFrame(_af_rows).set_index("Filter")

_sep  = "=" * 74
_sep2 = "-" * 74
print(_sep)
print("  ADVANCED FILTER COMPARISON — Win Rate / Total R / Max DD")
print(_sep)
display(_af_result)

# ─────────────────────────────────────────────────────────────────────────────
# Direction breakdown
# ─────────────────────────────────────────────────────────────────────────────
print()
print(_sep2)
print("  Direction breakdown (BUY vs SELL) for each filter")
print(_sep2)
_af_dir_rows = []
for _lbl, _mask in _af_filters:
    _sub = _t[_mask].reset_index(drop=True) if not isinstance(_mask, slice) else _t.copy()
    for _d in ["BUY", "SELL"]:
        _ds = _sub[_sub["direction"] == _d]
        _base_d = _t[_t["direction"] == _d]
        _n = len(_ds)
        _wr = (_ds["result"] == "TP").sum() / _n if _n > 0 else 0.0
        _r  = _ds["pnl_r"].sum() if _n > 0 else 0.0
        _af_dir_rows.append({
            "Filter":    _lbl,
            "Dir":       _d,
            "Trades":    _n,
            "Kept":      f"{_n/len(_base_d)*100:.0f}%",
            "Win Rate":  f"{_wr*100:.1f}%",
            "Total R":   f"{_r:+.1f}R",
        })
_af_dir_df = pd.DataFrame(_af_dir_rows)
display(_af_dir_df.set_index(["Filter", "Dir"]))

# ─────────────────────────────────────────────────────────────────────────────
# Equity curves
# ─────────────────────────────────────────────────────────────────────────────
print()
print(_sep2)
print("  Equity Curves — Cumulative R")
print(_sep2)

_af_palette = [
    ("#888888", "dash",  2),
    ("#9E9E9E", "dot",   1),
    ("#2196F3", "solid", 2),
    ("#FF9800", "solid", 2),
    ("#4CAF50", "solid", 2),
    ("#00BCD4", "dot",   2),
    ("#F44336", "solid", 2.5),
    ("#E91E63", "solid", 2.5),
]
_af_fig = go.Figure()
for (_lbl, _mask), (_clr, _dash, _w) in zip(_af_filters, _af_palette):
    _sub = _t[_mask].reset_index(drop=True) if not isinstance(_mask, slice) else _t.copy()
    if len(_sub) == 0: continue
    _cum = _sub["pnl_r"].cumsum()
    _af_fig.add_trace(go.Scatter(
        y=_cum, mode="lines", name=_lbl,
        line=dict(color=_clr, width=_w, dash=_dash),
    ))
_af_fig.add_hline(y=0, line_color="white", line_width=0.5, opacity=0.3)
_af_fig.update_layout(
    title="Advanced Filters — Equity Curves (Cumulative R, sorted by entry_time)",
    xaxis_title="Trade # (chronological)",
    yaxis_title="Cumulative R",
    template="plotly_dark",
    legend=dict(x=0.01, y=0.99, font_size=11),
    height=500,
)
_af_fig.show()

# ─────────────────────────────────────────────────────────────────────────────
# Findings
# ─────────────────────────────────────────────────────────────────────────────
print()
print(_sep)
print("  FINDINGS")
print(_sep)
_af_num = _af_result.copy()
_af_num["_r"]  = _af_num["Total R"].str.replace("R","").str.replace("+","").astype(float)
_af_num["_wr"] = _af_num["Win Rate"].str.replace("%","").replace("—","0").astype(float)
_af_num["_dd"] = _af_num["Max DD"].str.replace("R","").astype(float)
_af_num["_ex"] = _af_num["Expectancy"].str.replace("+","").replace("—","0").astype(float)
for _col, _lbl_col, _best_fn in [
    ("_wr", "Win Rate",   "idxmax"),
    ("_r",  "Total R",    "idxmax"),
    ("_dd", "Max DD",     "idxmax"),
    ("_ex", "Expectancy", "idxmax"),
]:
    _idx = getattr(_af_num[_col], _best_fn)()
    _val = _af_result.loc[_idx, _lbl_col]
    _flag = " ← BEST" if _idx != "Baseline (no filter)" else ""
    print(f"  Best {_lbl_col:12s}: {_idx}  →  {_val}{_flag}")

print()
print("  Baseline reference:")
base = "Baseline (no filter)"
for _col in ["Trades", "Win Rate", "Total R", "Max DD", "Expectancy"]:
    print(f"    {_col:12s}: {_af_result.loc[base, _col]}")
````

## 14. SL Failure Diagnostics → Targeted Filters

**Goal:** Inspect *only the losing trades* (SL hits) to learn *why* they fail,
then design filters that drop the conditions producing those losses.

### Step 1 — Compare TP vs SL across context features
For every potential filter input we compute `TP-rate` (a.k.a. win-rate) within
each bucket so we can spot the buckets where the losers cluster.

### Step 2 — Build targeted filters from the failure clusters

| # | Filter | Why it might work (failure pattern it removes) |
|---|--------|------------------------------------------------|
| H1 | **Drop stale OB (age > 2 H1 bars)** | Stale OBs (~4h+ from formation) lose at 70% rate |
| H2 | **Drop wide-risk (risk/ATR > Q75)** | Wide stops lose at 65% rate — big OB = exhaustion |
| H3 | **Drop H1 RSI < 30** | Extreme-RSI counter-momentum buys/sells lose at 92% rate |
| H4 | **Drop NY session** | NY underperforms (45.9% WR vs 50–54% in Asia/London) |
| H5 | **Combo: fresh OB + RSI ≥ 30 + narrow risk** | Stack the top 3 single filters |
| H6 | **Combo: H5 + D1 trend aligned** | Add macro trend gate for further selectivity |
| H7 | **Combo: H5 + no NY** | Add session gate without forcing D1 alignment |

All higher-TF indicators are looked up **lookahead-free** (last fully-closed
bar before each trade's `entry_time`).

````python
# ─────────────────────────────────────────────────────────────────────────────
# 14. SL Failure Diagnostics → Targeted Filters
# ─────────────────────────────────────────────────────────────────────────────

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

_SD_PERIOD   = 14
_SD_RISK_PCT = 1.0

# ─────────────────────────────────────────────────────────────────────────────
# Load trades + ohlcv (standalone — does not depend on prior cells)
# ─────────────────────────────────────────────────────────────────────────────
_sd_csv = Path("results/08_order_block_reaction/XAUUSD/single_pos/trades_raw.csv")
_sd_t   = pd.read_csv(_sd_csv)
_sd_t   = _sd_t[_sd_t["result"].isin(["TP", "SL"])].copy()
_sd_t["entry_time"]   = pd.to_datetime(_sd_t["entry_time"])
_sd_t["ob_time"]      = pd.to_datetime(_sd_t["ob_time"])
_sd_t["retest_time"]  = pd.to_datetime(_sd_t["retest_time"])
_sd_t                 = _sd_t.sort_values("entry_time").reset_index(drop=True)
_sd_t["risk"]         = (_sd_t["entry_price"] - _sd_t["sl"]).abs()
_sd_t["ob_age_h"]     = (_sd_t["entry_time"] - _sd_t["ob_time"]).dt.total_seconds() / 3600
_sd_t["hour"]         = _sd_t["entry_time"].dt.hour
_sd_t["session"]      = _sd_t["hour"].apply(lambda h: "Asia" if h < 14 else ("London" if h < 19 else "NY"))

# Indicator helpers
def _sd_ema(s, n): return s.ewm(span=n, adjust=False).mean()
def _sd_atr(df, n=14):
    h, l, c = df["high"], df["low"], df["close"]
    pc = c.shift(1)
    tr = pd.concat([(h - l), (h - pc).abs(), (l - pc).abs()], axis=1).max(axis=1)
    return tr.ewm(span=n, adjust=False).mean()
def _sd_rsi(s, n=14):
    d = s.diff(); g = d.clip(lower=0); l = (-d).clip(lower=0)
    ag = g.ewm(com=n-1, adjust=False).mean()
    al = l.ewm(com=n-1, adjust=False).mean()
    return 100 - (100 / (1 + ag / al.replace(0, np.nan)))

# Merge H1 + D1 indicators (lookahead-free)
for _tf, _delta in [("h1", pd.Timedelta(hours=1)), ("d1", pd.Timedelta(days=1))]:
    _o = pd.read_csv(Path(f"data/XAUUSD/{_tf.upper()}/ohlcv.csv"))
    _o["time"] = pd.to_datetime(_o["time"])
    _o["ema14"] = _sd_ema(_o["close"], _SD_PERIOD)
    _o["atr14"] = _sd_atr(_o,           _SD_PERIOD)
    _o["rsi14"] = _sd_rsi(_o["close"],  _SD_PERIOD)
    _o["close_time"] = _o["time"] + _delta
    _cols = ["close_time", "close", "ema14", "atr14", "rsi14"]
    _sub = _o[_cols].rename(columns={c: f"{_tf}_{c}" for c in _cols if c != "close_time"})
    _sub = _sub.rename(columns={"close_time": "entry_time"}).sort_values("entry_time")
    _sd_t = pd.merge_asof(_sd_t.sort_values("entry_time"), _sub,
                          on="entry_time", direction="backward")

_sd_t["risk_atr"]   = _sd_t["risk"] / _sd_t["h1_atr14"]
_sd_t["d1_aligned"] = ((_sd_t["direction"] == "BUY")  & (_sd_t["d1_close"] > _sd_t["d1_ema14"])) | \
                      ((_sd_t["direction"] == "SELL") & (_sd_t["d1_close"] < _sd_t["d1_ema14"]))

# ─────────────────────────────────────────────────────────────────────────────
# Step 1 — Diagnostic table: where do the LOSERS cluster?
# ─────────────────────────────────────────────────────────────────────────────
_sep  = "═" * 92
_sep2 = "─" * 92

def _sd_bucket_stats(df, col, label, bins=None, qcut_n=None, labels=None):
    if bins is not None:
        df = df.assign(_b=pd.cut(df[col], bins))
    elif qcut_n is not None:
        df = df.assign(_b=pd.qcut(df[col], qcut_n, labels=labels, duplicates="drop"))
    else:
        df = df.assign(_b=df[col])
    rows = []
    for v, g in df.groupby("_b", observed=True):
        n  = len(g)
        if n == 0: continue
        tp = int((g["result"] == "TP").sum())
        sl = int((g["result"] == "SL").sum())
        wr = tp / n * 100
        rows.append({
            "Feature": label, "Bucket": str(v),
            "n": n, "TP": tp, "SL": sl,
            "WR": f"{wr:5.1f}%", "Net R": f"{g['pnl_r'].sum():+.1f}",
        })
    return rows

_sd_diag = []
_sd_diag += _sd_bucket_stats(_sd_t, "session",  "Session")
_sd_diag += _sd_bucket_stats(_sd_t, "direction", "Direction")
_sd_diag += _sd_bucket_stats(_sd_t, "ob_age_h",  "OB Age (h)",
                              qcut_n=4, labels=["fresh", "recent", "old", "stale"])
_sd_diag += _sd_bucket_stats(_sd_t, "risk_atr",  "Risk / H1 ATR",
                              qcut_n=4, labels=["narrow", "mid-low", "mid-high", "wide"])
_sd_diag += _sd_bucket_stats(_sd_t, "h1_rsi14",  "H1 RSI(14)",
                              bins=[0, 30, 40, 50, 60, 70, 100])
_sd_diag += _sd_bucket_stats(_sd_t, "h1_atr14",  "H1 ATR(14)",
                              qcut_n=4, labels=["low", "mid-low", "mid-high", "high"])
_sd_diag += _sd_bucket_stats(_sd_t, "d1_aligned", "D1 Trend (price vs EMA14)")

_sd_diag_df = pd.DataFrame(_sd_diag).set_index(["Feature", "Bucket"])

print(_sep)
print("  SL FAILURE DIAGNOSTICS — where do losing trades cluster?")
print(_sep)
print(f"  Source : {_sd_csv}")
print(f"  Closed : {len(_sd_t)}   TP: {(_sd_t.result=='TP').sum()}   "
      f"SL: {(_sd_t.result=='SL').sum()}   Baseline WR: "
      f"{(_sd_t.result=='TP').mean()*100:.1f}%")
print(_sep2)
display(_sd_diag_df)

# ─────────────────────────────────────────────────────────────────────────────
# Step 2 — Build filters from the failure patterns and re-evaluate
# ─────────────────────────────────────────────────────────────────────────────
# Define what counts as "fresh" / "narrow risk" — drop the worst quartiles only
_sd_age_thr  = _sd_t["ob_age_h"].quantile(0.75)   # drop stale (top quartile)
_sd_risk_thr = _sd_t["risk_atr"].quantile(0.75)   # drop wide-risk (top quartile)

_mH1 = _sd_t["ob_age_h"]  <= _sd_age_thr             # fresh / non-stale OB
_mH2 = _sd_t["risk_atr"]  <= _sd_risk_thr            # narrow / non-wide stop
_mH3 = _sd_t["h1_rsi14"]  >= 30                      # not extremely oversold
_mH4 = _sd_t["session"]   != "NY"                    # exclude NY session
_mH5 = _mH1 & _mH2 & _mH3                            # combo top-3
_mH6 = _mH5 & _sd_t["d1_aligned"]                    # + D1 trend
_mH7 = _mH5 & _mH4                                   # + no NY

_sd_filters = [
    ("Baseline (no filter)",                    pd.Series(True, index=_sd_t.index)),
    ("H1: Drop stale OB",                       _mH1),
    ("H2: Drop wide-risk",                      _mH2),
    ("H3: Drop RSI<30",                         _mH3),
    ("H4: Drop NY session",                     _mH4),
    ("H5: Fresh + RSI>=30 + Narrow",            _mH5),
    ("H6: H5 + D1 aligned",                     _mH6),
    ("H7: H5 + no NY",                          _mH7),
]

def _sd_metrics(df, mask):
    g = df[mask].sort_values("entry_time").reset_index(drop=True)
    n = len(g)
    if n == 0:
        return dict(Trades=0, **{"Win Rate":"—","Total R":"+0.0R","Max DD":"0.0R",
                                 "Expectancy":"—","Return %":"+0.0%"})
    tp = int((g["result"] == "TP").sum())
    wr = tp / n * 100
    tot = g["pnl_r"].sum()
    eq  = g["pnl_r"].cumsum()
    dd  = (eq - eq.cummax()).min()
    exp = tot / n
    ret = tot * _SD_RISK_PCT
    return dict(Trades=n, **{
        "Win Rate":   f"{wr:5.1f}%",
        "Total R":    f"{tot:+.1f}R",
        "Max DD":     f"{dd:+.1f}R",
        "Expectancy": f"{exp:+.3f}",
        "Return %":   f"{ret:+.1f}%",
    })

_sd_rows = {lbl: _sd_metrics(_sd_t, m) for lbl, m in _sd_filters}
_sd_result = pd.DataFrame(_sd_rows).T

print()
print(_sep)
print("  TARGETED FILTER RESULTS (built from the diagnostic patterns above)")
print(_sep)
print(f"  fresh threshold     : ob_age_h  <= {_sd_age_thr:5.2f} h   (drop top quartile)")
print(f"  narrow-risk thresh  : risk/ATR  <= {_sd_risk_thr:5.2f}     (drop top quartile)")
print(f"  account risk/trade  : {_SD_RISK_PCT:.1f}%")
print(_sep2)
display(_sd_result)

# ─────────────────────────────────────────────────────────────────────────────
# Equity curves for every filter
# ─────────────────────────────────────────────────────────────────────────────
print()
print(_sep2)
print("  Equity Curves — Cumulative R")
print(_sep2)

_sd_palette = [
    ("#888888", "dash",  2),
    ("#FFC107", "solid", 2),
    ("#03A9F4", "solid", 2),
    ("#8BC34A", "solid", 2),
    ("#FF5722", "solid", 2),
    ("#E91E63", "solid", 2.5),
    ("#9C27B0", "solid", 2.5),
    ("#00BCD4", "solid", 2.5),
]
_sd_fig = go.Figure()
for (_lbl, _mask), (_clr, _dash, _w) in zip(_sd_filters, _sd_palette):
    _sub = _sd_t[_mask].sort_values("entry_time").reset_index(drop=True)
    if len(_sub) == 0: continue
    _cum = _sub["pnl_r"].cumsum()
    _sd_fig.add_trace(go.Scatter(
        y=_cum, mode="lines", name=f"{_lbl} (n={len(_sub)})",
        line=dict(color=_clr, width=_w, dash=_dash),
    ))
_sd_fig.add_hline(y=0, line_color="white", line_width=0.5, opacity=0.3)
_sd_fig.update_layout(
    title="SL-Diagnostic Filters — Equity Curves (Cumulative R)",
    xaxis_title="Trade # (chronological)",
    yaxis_title="Cumulative R",
    template="plotly_dark",
    legend=dict(x=0.01, y=0.99, font_size=10),
    height=520,
)
_sd_fig.show()

# ─────────────────────────────────────────────────────────────────────────────
# Findings & ranking
# ─────────────────────────────────────────────────────────────────────────────
print()
print(_sep)
print("  FINDINGS")
print(_sep)
_sd_num = _sd_result.copy()
_sd_num["_wr"] = _sd_num["Win Rate"].astype(str).str.replace("%","").replace("—","0").astype(float)
_sd_num["_r"]  = _sd_num["Total R"].astype(str).str.replace("R","").str.replace("+","").astype(float)
_sd_num["_dd"] = _sd_num["Max DD"].astype(str).str.replace("R","").astype(float)
_sd_num["_ex"] = _sd_num["Expectancy"].astype(str).str.replace("+","").replace("—","0").astype(float)

for _col, _lbl, _fn in [("_wr","Win Rate","idxmax"), ("_r","Total R","idxmax"),
                         ("_dd","Max DD","idxmax"),  ("_ex","Expectancy","idxmax")]:
    _idx = getattr(_sd_num[_col], _fn)()
    _val = _sd_result.loc[_idx, _lbl]
    _flag = " ← BEST" if _idx != "Baseline (no filter)" else ""
    print(f"  Best {_lbl:11s}: {_idx}  →  {_val}{_flag}")

print()
print("  Baseline reference:")
for _col in ["Trades","Win Rate","Total R","Max DD","Expectancy","Return %"]:
    print(f"    {_col:11s}: {_sd_result.loc['Baseline (no filter)', _col]}")
````
